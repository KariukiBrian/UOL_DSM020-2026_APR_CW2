# 1. Introduction and Literature Review

Cryptocurrency markets provide an interesting setting for data analysis as prices, trading activity, blockchain activity, and public information can all change rapidly. In contrast to typical financial markets, cryptocurrencies additionally provide community blockchain records that may be consolidated with market and media data (Koutmos, 2018). This provides a chance to examine market behaviour not just on price, but from a number of angles. Trading volume, blockchain transaction activity, and public attention have been found to carry significant insights regarding cryptocurrency behavior in previous research, with the majority of studies focusing on Bitcoin rather than XRP (Balcilar et al., 2017; Al Guindy, 2021). This project focuses specifically on XRP and combines market, XRP Ledger network, and media sentiment data in a reproducible Python workflow. Its intent is not to build an investment strategy, but to determine if measurable evidence of market activity, network activity, and public sentiment are correlated with XRP returns and volatility.

## 1.1 XRP Market and XRPL Context

XRP is a major cryptocurrency and the native digital asset of the XRP Ledger (XRPL). It is traded on cryptocurrency trading platforms, and its transactions are recorded on the XRP Ledger. These two sources of information provide different views of activity. Market data refer to variables like price and trading volume, while network data refer to activity directly on the blockchain. Previous cryptocurrency research suggests that trading activity may be related to price behaviour. For instance, Balcilar et al. (2017) investigated the relationship between Bitcoin trading volume and returns and concluded that trading volume can offer information on returns over parts of the return distribution. The same result was obtained by Aalborg, Molnár and de Vries (2019) when they built a model to explain the volatility of Bitcoin. These findings provide a reason to investigate whether XRP trading volume is similarly associated with the magnitude of daily XRP price movements.

Blockchain activity may provide additional information that is not captured by exchange trading volume. Koutmos (2018) discovered ties between the volume of transactions in Bitcoin and market returns, indicating that changes in transaction volume and market activity are linked. Gkillas, Tantoula and Tzagarakis (2021) similarly reported that Bitcoin transaction activity contained useful information when forecasting realised volatility. However, relationships identified for Bitcoin cannot automatically be assumed to apply to XRP because the assets have different network structures, users, and transaction patterns. XRP-specific research is less extensive, but Sankaewtong and Ikeda (2025) offer data pointing towards periods of XRP price fluctuation that correspond to structural changes in the transaction network. The results indicate that analyzing XRP Ledger transactions in conjunction with market information could yield further information about XRP market behavior. Therefore, the number of transactions processed on the XRP Ledger every day is used as a direct indicator of the network activity and is included in the present study.

## 1.2 Media Sentiment and Cryptocurrency Behaviour

The reaction of the investors and the public to the information can also affect cryptocurrency prices. Online news stories and social-media messages, among other types of online discussion, can evolve rapidly and can impact market focus, expectations, and trading decisions. In the study, Al Guindy (2021) analyzed around 25 million observations from social media over major cryptocurrencies and found a relationship between investor attention to cryptocurrencies and increased volatility in cryptocurrency prices. The study also identified a potential role for attention in providing information about future volatility. This helps substantiate the broader thesis that the market factors of cryptocurrency behavior are not just market dynamics, but also the level of public interest in an asset. Hence, media activity is relevant to the present project as there is a possibility that more media reports may be published when the market is experiencing unusual conditions, and the tone of the reports may provide information on the reactions of the public to XRP-related activities.

The correlation between sentiment and future returns is less clear. Guégan and Renault (2021) identified that the sentiment measured in social media at very high frequencies is statistically related to returns of Bitcoin, whereas the relationship was no longer observed with returns from lower frequencies. This may indicate that any sentiment effect could be temporary and that sentiment information across the day does not always contain large amounts of information. The present study therefore tests the relationship rather than assuming that positive or negative news predicts future XRP returns. Sentiment is calculated by the rule-based VADER method proposed by Hutto and Gilbert (2014) for headlines. VADER is beneficial because it yields a clear compound sentiment score and doesn't require the one-time creation of a separate training set to be labelled. It was meant for general social media language, though, not a language used in cryptocurrency trading. The sentiment scores used in this study are thus considered approximate rather than perfect measures of investors' beliefs about the tone of the headlines.

## 1.3 Research Gap

A significant gap in the current literature is the fact that the majority of empirical studies of cryptocurrency markets are conducted on Bitcoin. The trading volume, transaction activity, volatility, investor attention and sentiment have been studied extensively for Bitcoin (Balcilar et al., 2017; Koutmos, 2018; Guégan and Renault, 2021). While these studies offer valuable methods and expectations, they are not likely to be directly applicable to XRP. XRP has its own trading market and transaction network, and XRP-specific network research such as Sankaewtong and Ikeda (2025) suggests that changes in its on-chain structure can be related to price behaviour. It thus makes sense to look at XRP itself, rather than using evidence from Bitcoin as evidence of all cryptocurrencies. This project aims to fill in that void, but goes further with its specific focus on XRP as well as testing market, network and media variables in the same analytical study.

The second gap is related to the integration of various cryptocurrency information types. Current research tends to focus on trading activity, blockchain behaviour or sentiment, in isolation. This can be challenging to grasp the relationship between various types of activity and market behaviour in a single set of data and framework of analysis. The present project addresses this issue by combining daily market information, XRP Ledger transaction activity and XRP-related media headlines using a reproducible Python pipeline. It also introduces an exploratory XRP Market Attention and Stress Index (XMASI), which combines standardised measures from several data domains. The index is not considered predictive and is tested empirically for subsequent volatility. This way, they get the value-added, but it is clear that there is no certainty. The project thus makes its contribution primarily in terms of analysis that is specific to XRP, integration from multiple sources, reproducible data programming and critical evaluation, not by asserting that a new indicator is a reliable market predictor.

Research on cryptocurrency markets has produced evidence that:

However, much of the existing empirical literature concentrates on **Bitcoin**, while comparatively less research has examined XRP. In addition, market activity, blockchain activity and sentiment are often analysed separately.

## 1.5 Research Question

The main research question for this project is:

> To what extent are XRP trading activity, XRP Ledger network activity and public media sentiment associated with contemporaneous and subsequent XRP market behaviour during the study period?

The question is investigated through three related components:

- Is higher XRP trading activity associated with larger daily price movements?
- Is greater XRPL transaction activity associated with higher XRP market volatility?
- Is the sentiment of XRP-related media coverage associated with the following day's XRP return?

The first two questions examine relationships between market and network activity, while the third introduces a temporal element by examining whether media sentiment precedes subsequent price movement.

## 1.6 Research Hypotheses

Three hypotheses are defined before the main statistical analysis so that the study does not select relationships only after observing the results. The significance level is set at 5%. Each hypothesis uses variables that can be directly calculated from the collected data and that relate to previous cryptocurrency research. The first examines market trading activity, the second examines blockchain activity, and the third examines media sentiment. Together, these hypotheses allow the study to compare whether market, network and media information show similar or different relationships with XRP behaviour. The hypotheses test statistical association rather than causality. A significant result would therefore show that two variables move together in a systematic way within the observed sample, but it would not prove that one variable directly causes the other.

### Hypothesis 1: Trading Volume and Absolute Daily Returns

The first hypothesis examines whether greater XRP trading activity is associated with larger daily price movements. Previous evidence provides a basis for expecting a relationship between cryptocurrency volume and returns. Balcilar et al. (2017) found that Bitcoin trading volume could provide information about returns across parts of the return distribution, while Aalborg, Molnár and de Vries (2019) showed that trading volume contributed useful information when modelling Bitcoin volatility. In the present study, trading activity is measured using daily XRP trading volume in US dollars, while the magnitude of price movement is measured using absolute daily percentage return. Absolute return is used because the hypothesis concerns the size of a price movement rather than whether the price increased or decreased.



**Null hypothesis (Ho₁):**

> There is no statistically significant association between XRP daily trading volume and absolute daily return.

**Alternative hypothesis (H₁₁):**

> There is a statistically significant positive association between XRP daily trading volume and absolute daily return.

The variables are operationally defined as:

- Trading activity: daily XRP trading volume in USD.
- Price movement:absolute value of the daily percentage return.


### Hypothesis 2: XRPL Network Activity and Market Volatility

The second hypothesis examines whether activity on the XRP Ledger is associated with XRP market volatility. Previous blockchain research suggests that transaction activity can contain information about cryptocurrency market behaviour. Koutmos (2018) identified links between Bitcoin transaction activity and returns, while Gkillas, Tantoula and Tzagarakis (2021) found that Bitcoin transaction activity could improve forecasts of realised volatility. XRP-specific research also suggests that changes within the XRP transaction network can occur alongside important price movements (Sankaewtong and Ikeda, 2025). In this study, network activity is measured using the daily number of XRP Ledger transactions, while market volatility is measured using the seven-day rolling standard deviation of daily XRP returns. Because previous research does not establish that the direction must be the same for XRP as for Bitcoin, the alternative hypothesis is intentionally non-directional.

**Null hypothesis (H₀₂):**

> There is no statistically significant association between XRPL transaction activity and XRP rolling market volatility.

**Alternative hypothesis (H₁₂):**

> There is a statistically significant association between XRPL transaction activity and XRP rolling market volatility.

The variables are operationally defined as:

- Network activity: mean transaction count across systematically sampled validated XRP Ledgers for each day.
- Market volatility: seven-day rolling standard deviation of XRP daily returns.


### Hypothesis 3: Media Sentiment and Subsequent XRP Returns

The third hypothesis investigates whether the tone of XRP-related media coverage is associated with the following day's XRP return. This relationship is treated cautiously because previous evidence shows that sentiment effects can depend strongly on the frequency of the data. Guégan and Renault (2021) found that social-media sentiment was associated with Bitcoin returns at very high frequencies, but the relationship was not maintained at lower frequencies. Al Guindy (2021) also demonstrates that cryptocurrency attention can be related to market volatility, showing that public information and market behaviour may interact even when direct return prediction is difficult. In the present study, daily media sentiment is calculated as the mean VADER compound score of available XRP headlines, following the sentiment framework developed by Hutto and Gilbert (2014). This daily sentiment measure is compared with the next calendar day's XRP return.

**Null hypothesis (H₀₃):**

> Daily XRP-related media sentiment has no statistically significant association with the following day's XRP return.

**Alternative hypothesis (H₁₃):**

> Daily XRP-related media sentiment has a statistically significant association with the following day's XRP return.

The variables are operationally defined as:

- Media sentiment: mean daily sentiment score calculated from XRP-related headlines using a pre-trained sentiment-analysis method.
- Subsequent market return: XRP percentage return on the following calendar day.

### 1.7.1 Data Sources

Three independent public sources are used. 
-	**Market Data** - XRP market data are obtained from CoinGecko and include daily price, market capitalisation and trading volume measured in US dollars. CoinGecko is well suited as it offers structured historical cryptocurrency market data that can be accessed and downloaded programmatically and locally for the sake of reproducibilities. 
-	**Network Data** - The data on the XRP Ledger Network comes from the Coin Metrics public data archive. The main network variable is ‘TxCnt’ which denotes the number of transactions that are logged daily on the network. 
-	**Media Data** - Media data are obtained from the public Crypto News Dataset maintained by Speirsy11, from which XRP news records are selected using the article headline, publication date, source and related metadata. In this dataset, sentiment fields that already have information are not included in the analysis and instead sentiment is calculated from scratch with VADER. VADER was developed by Hutto and Gilbert (2014) as a transparent rule-based approach to sentiment analysis and provides a compound score ranging from negative to positive sentiment.

The three sources help to mitigate dependency on a single source and provide three lenses to view the research question: the market, the blockchain and the media. All data sources are first saved as raw data and then transformed. This separation is crucial as it allows the original observations to be stored for audit and comparison with the cleaned versions. Some data were collected from live data sources (APIs and public repositories) during development, but the final reproducible workflow relies on locally stored raw extracts where appropriate. This means that the assessor does not need to hold private API credentials, nor rely on external services being available during the execution of the notebook. The API key used for CoinGecko is not saved inside the final notebook as it was used during the development process.

### 1.7.2 Study Period

The overall market study period is from 15 Sept 2025 to 31 August 2026. This is a period with 351 observations of the XRP market every day, which is the primary periodicity for the project. The three independent data sources do not, however, have the same coverage. The number of required XRP transaction measures for the Coin Metrics network data can be found for the period from 15 September 2025 to 23 May 2026; 251 network observations are generated prior to any derived rolling measures. Media coverage is significantly less common with valid daily sentiment on 63 dates during the larger study interval. The study does not therefore simulate network and sentiment observations on dates when there is no data in the relevant source.

For this reason, individual statistical tests are based on observations available for each of the research questions. The market data is used with H1 and has the widest coverage. H2 is based on dates for which both network activity and the required volatility measure are known. H3 is based on dates that have an observed daily sentiment score and have a valid XRP return for the following day. This method results in unequal sample sizes between hypotheses and doesn't count missing information as observed information. A day with no XRP headline does not necessarily mean that the sentiment is negative. The variation in source coverage is reported where necessary, throughout the analysis, and is considered a limitation when interpreting the results.

### 1.7.3 Variables

The study builds variables directly corresponding to the research hypotheses. Variables are divided into four groups: market variables, network variables, media variables and engineered analytical variables.
#### Market Variables

The original market variables are:

- `price_usd` — daily XRP price in US dollars;
- `market_cap_usd` — daily XRP market capitalisation;
- `market_volume_usd` — daily XRP trading volume.

The following features are derived from these variables:

- `daily_return` — percentage change in XRP price from the previous day;
- `absolute_return` — absolute value of the daily return;
- `rolling_volatility_7d` — seven-day rolling standard deviation of daily returns;
- `rolling_volatility_30d` — thirty-day rolling standard deviation of daily returns;
- `volume_change` — daily percentage change in trading volume;
- `log_market_volume` — logarithmic transformation of trading volume;
- `volume_zscore` — standardised trading-volume measure.
#### Network Variables

The original XRPL variables are:

- `ledger_index` — unique index identifying a validated ledger;
- `close_time` — UTC time at which the ledger was validated;
- `transaction_count` — number of transactions contained in the sampled ledger.

After aggregation, the main network variables are:

- `mean_transaction_count` — mean number of transactions across sampled ledgers for that day;
- `network_activity_change` — change in network activity from the previous day;
- `network_activity_zscore` — standardised network-activity measure.

#### Media Variables

The original media variables are:

- `published_at` — publication timestamp;
- `headline` — article headline;
- `source` — news source or domain;
- `url` — article URL.

The derived media variables are:

- `sentiment_score` — numerical sentiment score for an individual headline;
- `mean_sentiment` — mean headline sentiment for each day;
- `median_sentiment` — median headline sentiment for each day;
- `article_count` — number of XRP-related articles observed each day;
- `positive_ratio` — proportion of daily headlines classified as positive;
- `negative_ratio` — proportion of daily headlines classified as negative;
- `sentiment_7d_ma` — seven-day moving average of daily sentiment.

#### Lagged Variables

To introduce a temporal component to the analysis, the following variables are created:

- `sentiment_lag_1d` — previous day's media sentiment;
- `network_activity_lag_1d` — previous day's network activity;
- `next_day_return` — XRP return on the following day.

#### Exploratory Composite Variable
The project also creates an exploratory XRP Market Attention and Stress Index (XMASI). The index is made up of standardised absolute return, logarithmic trading volume, logarithmic XRP Ledger transactions, logarithmic number of media articles and reversed headline sentiment. Inverting the sentiment component translates to an increase in negative sentiment having a positive impact on the stress measure. Equal weighting is employed for the sake of transparency and not using the same small dataset that was used for evaluation purposes. The computations of XMASI are performed only if all five components are known: if any of them is missing, it will be silently omitted on the respective dates, resulting in a different meaning of the index.


### 1.7.4 Statistical Approach

The statistical analysis starts with descriptive statistics and visual exploration of the main variables. Means, medians, standard deviations, quartiles and observed ranges are used to describe the datasets, while time-series plots and a Spearman correlation matrix are used to identify broad patterns. The skewness and Shapiro-Wilk test are then used to evaluate the distributional characteristics. The three primary variables exhibit significant skewness, outlying values and statistically significant deviations from normality. For this reason, the three main hypotheses are tested using **Spearman rank correlation** rather than Pearson correlation. Spearman correlation is suitable because it assumes monotonic relationships to be evaluated and it is less sensitive to non-normality and extreme values.

All principal hypothesis tests use a significance level of \(\alpha = 0.05\). H1 looks at the relationship between daily volume and absolute daily return. H2 looks at XRP Ledger transaction count as well as 7 days rolling volatility. H3 looks at the mean daily headline sentiment and subsequent day's XRP return. The correlation coefficient and the p-value are complemented by 95% bootstrap confidence intervals, which are calculated by using 2,000 resamples to give further information about the uncertainty of each effect estimate. The statistical significance is not considered as evidence of causality and the strength and direction of the estimated associations is interpreted in conjunction with the p-value.

A supplementary lag analysis is conducted on relationships with sentiment and return between -7 and +7 days. Since testing multiple lags increases the likelihood of finding a significant result by chance, Bonferroni adjusted p-values are computed for the 15 comparisons of lags. Spearman correlation and descriptive comparison of high-XMASI observations with the rest of the sample are used to assess the exploratory XMASI measure against volatility over the next 7 days. The comparisons of raw and log trading volume, seven-day and fourteen-day volatility, mean and median sentiment, and leave-one-component-out versions of XMASI are used in robustness checks. These tests are used to assess the extent to which the conclusions are sensitive to single-point decisions.These tests are not intended to pursue statistical significance but to determine if the results are sensitive to single-point decisions.

### 1.7.6 Ethical Considerations

The project relies solely on public information from existing markets and blockchain, and does not gather personal information directly from people. XRP Ledger activity is only analysed as a daily aggregated amount of transactions. Individual wallet addresses are not identified, profiled or linked to real people. Likewise, the analysis of the media is based on article headlines and the information given in the metadata of the publication and not on private communications. Making these decisions helps to minimise concerns about privacy and ensures that the project is focused on the behaviour of the aggregate market and not individual users.

However, there are some critical ethical and interpretive considerations. Data that are available to the public does not necessarily imply that they are unbiased. News datasets might over-represent specific publishers, events or languages while missing news headlines might not be because there isn't public discussion, but because of collection limitations. Sarcasm, financial terminology or context, and lack of specific development for XRP or cryptocurrency markets can result in sentiment models misinterpreting the sentiment (Hutto and Gilbert, 2014). It is therefore not the sentiment score itself that is treated as a direct indicator of investor sentiment.

The analysis also doesn't present relationships in statistics as financial "advice". Correlation does not imply causation and a history of association does not mean that this association will continue to exist in the future under the new market conditions. This is especially crucial when dealing with cryptocurrency markets, where prices can fluctuate significantly, and events outside of the dataset may impact market conditions. The exploratory XMASI measure is thus critically assessed and not offered as a trading signal. The project treats the following principles as ethical as much as possible: reproducibility, clear communication of insignificant results, and acknowledgment of missing data and differentiation of association, prediction and causation.

## 2.1 Data Acquisition and HTTP Functions

A small HTTP helper is defined for programmatic data acquisition. It sets a timeout, tests the HTTP response code and retries temporary failures before raising an informative error. It comes in handy for live API acquisition while developing. The notebook is, however, mostly based on saved raw datasets, so that the notebook can be reproduced without needing to have API credentials or continuous access to external services. The acquisition function is therefore not part of the analytical pipeline and won't alter the data downloaded.


```python
def request_json(
    url: str,
    method: str = "GET",
    params: Optional[Dict] = None,
    payload: Optional[Dict] = None,
    headers: Optional[Dict] = None,
    retries: int = 3,
    timeout: int = 30
) -> Dict:
    """Send an HTTP request and return a JSON response."""

    for attempt in range(1, retries + 1):

        try:

            response = requests.request(
                method=method,
                url=url,
                params=params,
                json=payload,
                headers=headers,
                timeout=timeout
            )

            if response.status_code == 429:

                wait = int(
                    response.headers.get(
                        "Retry-After",
                        2 ** attempt
                    )
                )

                logging.warning(
                    "Rate limit reached. Waiting %s seconds.",
                    wait
                )

                time.sleep(wait)
                continue

            response.raise_for_status()

            return response.json()

        except (
            requests.RequestException,
            ValueError
        ) as error:

            if attempt == retries:

                logging.error(
                    "Request failed after %s attempts.",
                    retries
                )

                raise

            wait = 2 ** (attempt - 1)

            logging.warning(
                "Attempt %s failed: %s. Retrying in %s seconds.",
                attempt,
                error,
                wait
            )

            time.sleep(wait)

    raise RuntimeError(
        "Request failed."
    )
```

```python

XRPL_URL = "https://s2.ripple.com:51234/"

def xrpl_request(
    method: str,
    params: Dict
) -> Dict:
    """Send a JSON-RPC request to the XRP Ledger."""

    payload = {
        "method": method,
        "params": [params]
    }

    response = request_json(
        XRPL_URL,
        method="POST",
        payload=payload
    )

    result = response.get("result", {})

    if result.get("status") != "success":
        raise ValueError(
            f"XRPL request failed: {result}"
        )

    return result
```

## 2.2 Data Validation and Audit Functions

Validation is used to check that the structure of the dataset is not unexpected before potentially large transformations. The first function ensures that all the fields necessary for that specific step exist. A second function generates a short audit of the size of the dataset, missing data, temporal coverage, and duplicate row information. These checks can be used to identify issues that may have occurred in the source data, as well as any transformations that have been applied during the cleaning process. Several other assertions are made later in the notebook to validate particular assumptions, like transaction count is non-negative or ranges of valid sentiment-scores.


```python
def validate_columns(
    df: pd.DataFrame,
    required_columns,
    dataset_name: str
) -> None:
    """
    Check that all required columns are present.
    """

    missing_columns = (
        set(required_columns)
        - set(df.columns)
    )

    if missing_columns:
        raise ValueError(
            f"{dataset_name} is missing required columns: "
            f"{sorted(missing_columns)}"
        )
```

```python
def audit_dataframe(
    df: pd.DataFrame,
    dataset_name: str,
    date_column: str = None
) -> pd.DataFrame:
    """
    Create a simple audit summary for a DataFrame.
    """

    earliest_date = None
    latest_date = None

    if date_column is not None and date_column in df.columns:

        dates = pd.to_datetime(
            df[date_column],
            errors="coerce",
            utc=True
        )

        if dates.notna().any():
            earliest_date = dates.min()
            latest_date = dates.max()

    return pd.DataFrame({
        "dataset": [dataset_name],
        "rows": [len(df)],
        "columns": [df.shape[1]],
        "missing_values": [
            int(df.isna().sum().sum())
        ],
        "duplicate_rows": [
            int(df.duplicated().sum())
        ],
        "earliest_date": [
            earliest_date
        ],
        "latest_date": [
            latest_date
        ]
    })
```

```python
def data_quality_summary(
    df: pd.DataFrame,
    dataset_name: str
) -> pd.DataFrame:
    """Return a concise data-quality summary."""

    return pd.DataFrame({
        "dataset": [dataset_name],
        "rows": [len(df)],
        "columns": [len(df.columns)],
        "missing_values": [int(df.isna().sum().sum())],
        "duplicate_rows": [int(df.duplicated().sum())]
    })
```

## 2.3 Cleaning and Text-Preparation Functions

The market and network and media datasets are different in structure and quality requirements, so that they are cleaned separately. The functions transform dates and numerical variables into a consistent format, exclude invalid observations, remove duplicate dates and filter the study period limits. Since each function starts with copying the original raw DataFrames, they are never modified. Media cleaning also normalizes the text in headlines, which helps to preserve the punctuation and case of the letters because they can affect the sentiment scores in VADER. No media or network observations are artificially generated in the process of cleaning.


```python
# Markett Data
def clean_market_data(
    df: pd.DataFrame
) -> pd.DataFrame:
    """Clean XRP market data."""

    required = {
        "date",
        "price_usd",
        "market_cap_usd",
        "market_volume_usd"
    }

    validate_columns(
        df,
        required,
        "Market data"
    )

    clean = df.copy()

    clean["date"] = pd.to_datetime(
        clean["date"],
        errors="coerce"
    )

    numeric = [
        "price_usd",
        "market_cap_usd",
        "market_volume_usd"
    ]

    clean[numeric] = clean[numeric].apply(
        pd.to_numeric,
        errors="coerce"
    )

    # Financial values outside valid bounds
    clean.loc[
        clean["price_usd"] <= 0,
        "price_usd"
    ] = np.nan

    clean.loc[
        clean["market_volume_usd"] < 0,
        "market_volume_usd"
    ] = np.nan

    clean.loc[
        clean["market_cap_usd"] < 0,
        "market_cap_usd"
    ] = np.nan

    clean = (
        clean
        .drop_duplicates(subset="date")
        .dropna(
            subset=[
                "date",
                "price_usd",
                "market_volume_usd"
            ]
        )
        .sort_values("date")
        .reset_index(drop=True)
    )

    return clean
```

```python
#network data (XRPL)

def clean_network_data(
    df: pd.DataFrame
) -> pd.DataFrame:
    """
    Clean daily XRP Ledger transaction data.
    """

    required = {
        "time",
        "TxCnt"
    }

    validate_columns(
        df,
        required,
        "Network data"
    )

    clean = df.copy()

    clean["time"] = pd.to_datetime(
        clean["time"],
        errors="coerce"
    )

    clean["TxCnt"] = pd.to_numeric(
        clean["TxCnt"],
        errors="coerce"
    )

    clean = clean.dropna(
        subset=[
            "time",
            "TxCnt"
        ]
    )

    clean = clean[
        clean["TxCnt"] >= 0
    ]

    clean = (
        clean
        .drop_duplicates(
            subset="time"
        )
        .rename(
            columns={
                "time": "date",
                "TxCnt": "transaction_count"
            }
        )
        .sort_values("date")
        .reset_index(drop=True)
    )

    return clean
```

```python
#Media Metadata

def clean_news_data(
    df: pd.DataFrame
) -> pd.DataFrame:
    """
    Clean XRP-related media observations.
    """

    required = {
        "published_at",
        "headline",
        "source_name"
    }

    validate_columns(
        df,
        required,
        "Media data"
    )

    clean = df.copy()

    clean["published_at"] = pd.to_datetime(
        clean["published_at"],
        errors="coerce",
        utc=True
    )

    clean["headline"] = (
        clean["headline"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    clean["source_name"] = (
        clean["source_name"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    clean = clean[
        clean["published_at"].notna()
        & clean["headline"].ne("")
    ]

    clean = clean.drop_duplicates(
        subset=[
            "published_at",
            "headline"
        ]
    )

    clean["date"] = (
        clean["published_at"]
        .dt.normalize()
        .dt.tz_localize(None)
    )

    clean = (
        clean
        .rename(
            columns={
                "source_name": "source"
            }
        )
        .sort_values("published_at")
        .reset_index(drop=True)
    )

    return clean
```

## 2.4 NLP and Sentiment Functions

Sentiment is measured independently from the sentiment variables provided by the original media dataset using VADER. VADER is a rule-based sentiment model that yields a sentiment score from -1 to +1, with increasingly negative scores indicating more negative sentiment and increasingly positive scores indicating more positive sentiment (Hutto and Gilbert, 2014). Ensuring consistency of headline across the media sample by using the same model. The standard VADER thresholds of 0.05 and -0.05 are used to create positive, neutral and negative descriptive labels. The continuous compound score remains the main variable used in the statistical analysis.


```python
#importing SentimentIntensityAnalyzer
%pip install vaderSentiment 
from vaderSentiment.vaderSentiment import (
    SentimentIntensityAnalyzer
)

sentiment_analyser = SentimentIntensityAnalyzer()
```

```python
#Adding VADER sentiment scores to media headlines

def add_sentiment(
    df: pd.DataFrame
) -> pd.DataFrame:
    """Add VADER sentiment scores to media headlines."""

    result = df.copy()

    result["clean_headline"] = (
        result["headline"]
        .str.replace(
            r"https?://\S+",
            "",
            regex=True
        )
        .str.replace(
            r"\s+",
            " ",
            regex=True
        )
        .str.strip()
    )

    result["sentiment_score"] = (
        result["clean_headline"]
        .apply(
            lambda text:
            sentiment_analyser
            .polarity_scores(text)["compound"]
        )
    )

    result["sentiment_label"] = np.select(
        [
            result["sentiment_score"] >= 0.05,
            result["sentiment_score"] <= -0.05
        ],
        [
            "positive",
            "negative"
        ],
        default="neutral"
    )

    return result
```

## 2.5 Feature Engineering and Transformation Functions

Feature engineering transforms cleaned observations to variables that can address the research questions. Standardisation is applied when several measurements have different numerical ranges, but must be compared. The features of the market are daily return, absolute return, rolling volatility, change in trading volume and logarithmic volume. The features on the network include transaction count change, transaction count shown in a logarithmic scale, a standardised activity measure and a 7 day moving average. To ensure that activity measures are strongly right skewed, logarithmic transformations are applied, but the original variables are still interpretable and used for hypothesis testing.


```python
# I will add Market features
def add_market_features(
    df: pd.DataFrame
) -> pd.DataFrame:
    """Create XRP return, volatility and volume features."""

    result = (
        df.copy()
        .sort_values("date")
        .reset_index(drop=True)
    )

    result["daily_return"] = (
        result["price_usd"]
        .pct_change(fill_method=None)
    )

    result["absolute_return"] = (
        result["daily_return"].abs()
    )

    result["rolling_volatility_7d"] = (
        result["daily_return"]
        .rolling(7)
        .std()
    )

    result["rolling_volatility_30d"] = (
        result["daily_return"]
        .rolling(30)
        .std()
    )

    result["log_market_volume"] = np.log1p(
        result["market_volume_usd"]
    )

    result["volume_zscore"] = stats.zscore(
        result["log_market_volume"],
        nan_policy="omit"
    )

    return result
```

```python
# Now on to Daily network and sentiment features

def aggregate_daily_data(
    network: pd.DataFrame,
    news: pd.DataFrame
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """Aggregate XRPL and media observations to daily level."""

    network_daily = network.copy()

    network_daily["date"] = (
        network_daily["close_time"]
        .dt.normalize()
        .dt.tz_localize(None)
    )

    network_daily = (
        network_daily
        .groupby("date", as_index=False)
        .agg(
            mean_transaction_count=(
                "transaction_count",
                "mean"
            ),
            sampled_ledgers=(
                "ledger_index",
                "count"
            )
        )
    )

    sentiment_daily = (
        news
        .groupby("date", as_index=False)
        .agg(
            mean_sentiment=(
                "sentiment_score",
                "mean"
            ),
            article_count=(
                "sentiment_score",
                "size"
            )
        )
    )

    return network_daily, sentiment_daily
```

```python
#Merge and lag variables
def merge_daily_data(
    market: pd.DataFrame,
    network: pd.DataFrame,
    sentiment: pd.DataFrame
) -> pd.DataFrame:
    """Combine the three daily datasets and add lagged variables."""

    merged = (
        market.copy()
        .merge(
            network,
            on="date",
            how="left"
        )
        .merge(
            sentiment,
            on="date",
            how="left"
        )
        .sort_values("date")
        .reset_index(drop=True)
    )

    merged["sentiment_lag_1d"] = (
        merged["mean_sentiment"]
        .shift(1)
    )

    merged["network_activity_lag_1d"] = (
        merged["mean_transaction_count"]
        .shift(1)
    )

    merged["next_day_return"] = (
        merged["daily_return"]
        .shift(-1)
    )

    return merged
```

## 2.6 Statistical Analysis Functions

The most important inferential procedure is the Spearman rank correlation. It is suitable for this project because the later distribution diagnostics show that several key variables are skewed, contain extreme observations and do not satisfy normality assumptions. The function below first removes observations where either member of a pair is missing and then calculates the correlation coefficient and p-value. A 95% bootstrap confidence interval is also estimated using 2,000 resamples. Providing an effect size along with a p-value and a confidence interval gives a more complete interpretation than just the p-value.


```python
#Correlation
def correlation_test(
    x: pd.Series,
    y: pd.Series,
    method: str = "spearman"
) -> Dict:
    """Run Pearson or Spearman correlation."""

    data = pd.concat(
        [x, y],
        axis=1
    ).dropna()

    if len(data) < 3:
        raise ValueError(
            "At least three paired observations are required."
        )

    if method == "spearman":
        coefficient, p_value = stats.spearmanr(
            data.iloc[:, 0],
            data.iloc[:, 1]
        )

    elif method == "pearson":
        coefficient, p_value = stats.pearsonr(
            data.iloc[:, 0],
            data.iloc[:, 1]
        )

    else:
        raise ValueError(
            "Method must be 'spearman' or 'pearson'."
        )

    return {
        "method": method,
        "n": len(data),
        "coefficient": coefficient,
        "p_value": p_value
    }
```

```python
#Bootstrap confidence interval
def bootstrap_correlation_ci(
    x: pd.Series,
    y: pd.Series,
    n_bootstrap: int = 2000,
    seed: int = 42
) -> Tuple[float, float]:
    """Estimate a 95% bootstrap CI for Spearman correlation."""

    data = pd.concat(
        [x, y],
        axis=1
    ).dropna()

    rng = np.random.default_rng(seed)

    estimates = []

    for _ in range(n_bootstrap):

        sample = data.iloc[
            rng.integers(
                0,
                len(data),
                len(data)
            )
        ]

        if (
            sample.iloc[:, 0].nunique() < 2
            or sample.iloc[:, 1].nunique() < 2
        ):
            continue

        rho = stats.spearmanr(
            sample.iloc[:, 0],
            sample.iloc[:, 1]
        ).statistic

        estimates.append(rho)

    return (
        float(np.percentile(estimates, 2.5)),
        float(np.percentile(estimates, 97.5))
    )
```

```python
#Lag analysis
def lagged_spearman(
    df: pd.DataFrame,
    x_column: str,
    y_column: str,
    max_lag: int = 7
) -> pd.DataFrame:
    """Calculate Spearman correlations across time lags."""

    results = []

    for lag in range(
        -max_lag,
        max_lag + 1
    ):

        data = pd.concat(
            [
                df[x_column].shift(lag),
                df[y_column]
            ],
            axis=1
        ).dropna()

        if len(data) < 3:
            continue

        rho, p_value = stats.spearmanr(
            data.iloc[:, 0],
            data.iloc[:, 1]
        )

        results.append({
            "lag_days": lag,
            "n": len(data),
            "rho": rho,
            "p_value": p_value
        })

    return pd.DataFrame(results)
```

Historical XRP market data are obtained from the **CoinGecko API** using the coin identifier `ripple`.

**1 September 2025 to 31 August 2026**

The API returns:

- XRP price in USD;
- market capitalisation in USD; and
- 24-hour trading volume in USD.

For periods longer than 90 days, CoinGecko provides daily observations, which are appropriate for the daily analytical design used in this project.

A CoinGecko Demo API key is read from an environment variable when live acquisition is enabled. The key is not written directly into the notebook.

```python
# Market data acquisition configuration

STUDY_START = "2025-09-15"
STUDY_END = "2026-08-31"

MARKET_RAW_PATH = RAW_DIR / "xrp_market_raw.csv"

# Keep False in the submitted notebook.
RUN_LIVE_MARKET_ACQUISITION = False
```

```python
# LLive CoinGecko acquisition.
# This block is disabled in the submitted notebook because the raw market dataset is supplied with the submission.

if RUN_LIVE_MARKET_ACQUISITION:

    COINGECKO_API_KEY = getpass(
        "Enter CoinGecko Demo API key: "
    )

    url = (
        "https://api.coingecko.com/api/v3/"
        "coins/ripple/market_chart/range"
    )

    start_timestamp = int(
        pd.Timestamp(
            STUDY_START,
            tz="UTC"
        ).timestamp()
    )

    end_timestamp = int(
        (
            pd.Timestamp(
                STUDY_END,
                tz="UTC"
            )
            + pd.Timedelta(days=1)
            - pd.Timedelta(seconds=1)
        ).timestamp()
    )

    data = request_json(
        url,
        params={
            "vs_currency": "usd",
            "from": start_timestamp,
            "to": end_timestamp
        },
        headers={
            "x-cg-demo-api-key": COINGECKO_API_KEY
        }
    )

    prices = pd.DataFrame(
        data["prices"],
        columns=[
            "timestamp",
            "price_usd"
        ]
    )

    market_caps = pd.DataFrame(
        data["market_caps"],
        columns=[
            "timestamp",
            "market_cap_usd"
        ]
    )

    volumes = pd.DataFrame(
        data["total_volumes"],
        columns=[
            "timestamp",
            "market_volume_usd"
        ]
    )

    market_raw = (
        prices
        .merge(
            market_caps,
            on="timestamp",
            how="outer"
        )
        .merge(
            volumes,
            on="timestamp",
            how="outer"
        )
    )

    market_raw["date"] = (
        pd.to_datetime(
            market_raw["timestamp"],
            unit="ms",
            utc=True
        )
        .dt.tz_localize(None)
    )

    market_raw = (
        market_raw[
            [
                "date",
                "price_usd",
                "market_cap_usd",
                "market_volume_usd"
            ]
        ]
        .sort_values("date")
        .reset_index(drop=True)
    )

    market_raw.to_csv(
        MARKET_RAW_PATH,
        index=False
    )

    print(
        "Live CoinGecko market data downloaded and saved."
    )
```

## 3.2 XRP Ledger Network Data

XRP Ledger network activity is obtained from the Coin Metrics public data archive. Coin Metrics publishes historical cryptocurrency network metrics in asset-level CSV files, including data for XRP. The data are accessed programmatically from the organisation's public GitHub repository, providing a reproducible alternative to repeatedly querying individual XRP Ledger nodes.

The principal network variable used in this study is `TxCnt`, which records the number of transactions occurring on the XRP Ledger during each daily interval. This provides a more complete representation of daily network activity than sampling individual ledgers at selected times. The measure represents transaction frequency and should not be interpreted as transaction value, the number of unique users, or the economic importance of the transactions.

An audit of the public XRP archive showed that 23 May 2026 was the latest date with a complete daily transaction-count observation available for the selected research period. The common analytical period was therefore defined as 15 September 2025 to 23 May 2026. Using a common period across the market, network and media datasets improves temporal comparability and avoids imputing unavailable network observations.

The complete source file is downloaded and stored locally before cleaning. This preserves the original source data and allows subsequent notebook executions to use exactly the same raw observations without depending on continued internet availability.

## 3.3 Media and News Data

Public media coverage relating to XRP is obtained from the GDELT DOC 2.0 API. GDELT provides a searchable archive of global online news and allows historical articles to be retrieved using specified start and end dates. Unlike the market-data source, the GDELT API used in this project does not require an authentication key, which simplifies reproducibility.

The search term `XRP` is used to identify articles associated with the cryptocurrency. The study period is divided into seven-day intervals rather than being submitted as a single request. This approach reduces the risk of reaching the maximum number of articles that the API can return for one request and provides more consistent coverage throughout the research period.

For each returned article, the publication time, headline, source domain, URL, language and source country are retained. Only article metadata and headlines are required for this project; full article text is not collected. Sentiment values supplied by external sources are not used because headline sentiment will be calculated independently using VADER later in the analytical pipeline.

The downloaded records are stored in their raw form before duplicate removal, language filtering or sentiment analysis. This maintains separation between data acquisition and data cleaning and allows the original media observations to be audited independently.

```python
# Media data acquisition configuration

NEWS_RAW_PATH = (
    RAW_DIR / "xrp_news_raw.csv"
)

REPO_DIR = (
    Path("data")
    / "source"
    / "crypto-news-dataset"
)

CRYPTO_NEWS_REPO = (
    "https://github.com/"
    "Speirsy11/"
    "crypto-news-dataset.git"
)

# Keep False in the submitted notebook.
RUN_LIVE_MEDIA_ACQUISITION = False
```

```python
# Optional live media acquisition.
# Disabled in the submitted notebook because
# the prepared XRP news dataset is supplied.

if RUN_LIVE_MEDIA_ACQUISITION:

    import os
    import shutil
    import subprocess

    # Check Git availability
    if shutil.which("git") is None:
        raise RuntimeError(
            "Git is required for live media acquisition."
        )

    try:
        subprocess.run(
            [
                "git",
                "lfs",
                "version"
            ],
            capture_output=True,
            text=True,
            check=True
        )

    except subprocess.CalledProcessError as error:
        raise RuntimeError(
            "Git LFS is required for live media acquisition."
        ) from error


    # Remove an incomplete previous clone
    if REPO_DIR.exists():
        shutil.rmtree(
            REPO_DIR
        )


    REPO_DIR.parent.mkdir(
        parents=True,
        exist_ok=True
    )


    # Initialise Git LFS
    subprocess.run(
        [
            "git",
            "lfs",
            "install"
        ],
        check=True
    )


    # Clone repository without downloading
    # all large Git LFS objects
    environment = os.environ.copy()

    environment[
        "GIT_LFS_SKIP_SMUDGE"
    ] = "1"


    print(
        "Downloading crypto-news dataset structure..."
    )


    subprocess.run(
        [
            "git",
            "clone",
            "--depth",
            "1",
            CRYPTO_NEWS_REPO,
            str(REPO_DIR)
        ],
        check=True,
        env=environment
    )


    print(
        "Downloading XRP news data..."
    )


    # Download only XRP news Parquet objects
    subprocess.run(
        [
            "git",
            "lfs",
            "pull",
            "--include="
            "sentiment/source_type=news/"
            "asset=XRP/**"
        ],
        cwd=REPO_DIR,
        check=True
    )


    # Find XRP Parquet files
    xrp_news_files = list(
        REPO_DIR.glob(
            "sentiment/"
            "source_type=news/"
            "asset=XRP/"
            "year=*/"
            "month=*/"
            "day=*/"
            "sentiment.parquet"
        )
    )


    if not xrp_news_files:
        raise FileNotFoundError(
            "No XRP news Parquet files were found."
        )


    # Retain only valid Parquet files
    valid_files = []

    for file in xrp_news_files:

        with open(
            file,
            "rb"
        ) as f:

            start = f.read(4)

            f.seek(
                -4,
                2
            )

            end = f.read(4)

        if (
            start == b"PAR1"
            and end == b"PAR1"
        ):
            valid_files.append(
                file
            )


    if not valid_files:
        raise ValueError(
            "No valid XRP Parquet files were found."
        )


    # Load article-level XRP news data
    news_parts = [
        pd.read_parquet(file)
        for file in valid_files
    ]

    news_source = pd.concat(
        news_parts,
        ignore_index=True
    )


    required_news_columns = [
        "document_id",
        "source_type",
        "source_name",
        "asset",
        "relevance_score",
        "headline",
        "url",
        "author",
        "published_at"
    ]


    validate_columns(
        news_source,
        required_news_columns,
        "XRP media source data"
    )


    # Select only fields required for this study.
    # Pre-existing sentiment fields are deliberately excluded.
    news_raw = (
        news_source[
            required_news_columns
        ]
        .copy()
    )


    news_raw["published_at"] = pd.to_datetime(
        news_raw["published_at"],
        errors="coerce",
        utc=True
    )


    # Restrict to the overall study period
    start_date = pd.Timestamp(
        STUDY_START,
        tz="UTC"
    )

    end_date = (
        pd.Timestamp(
            STUDY_END,
            tz="UTC"
        )
        + pd.Timedelta(days=1)
        - pd.Timedelta(seconds=1)
    )


    news_raw = news_raw[
        news_raw["published_at"].between(
            start_date,
            end_date
        )
    ].reset_index(drop=True)


    news_raw.to_csv(
        NEWS_RAW_PATH,
        index=False
    )


    print(
        "Live XRP media data downloaded and saved."
    )
```

# 4. Data Cleaning and Preparation

The three raw datasets are cleaned independently before they are transformed or merged. Cleaning is performed on copies of the raw DataFrames so that the original acquired observations remain unchanged throughout the project.

The cleaning process focuses on missing values, duplicate observations, inconsistent data types and values that fall outside reasonable validity rules. The effect of each cleaning stage is measured by comparing the number of observations before and after processing. No statistical features, sentiment scores or derived variables are created in this section; these are introduced later during data transformation and feature engineering.

Where possible, observations are corrected through appropriate type conversion rather than removed. Records are excluded only when essential variables cannot be interpreted reliably. This conservative approach reduces unnecessary information loss while maintaining a transparent and reproducible processing pipeline.

## 4.1 Market Data

The raw XRP market dataset contains daily price, market capitalisation and trading-volume observations obtained from CoinGecko. Before these variables can be used to calculate returns and volatility, the dataset is checked for missing values, duplicate dates, inappropriate data types and economically invalid numerical values.

Dates are converted to a consistent datetime representation and the financial variables are converted to numeric data types. XRP price and market capitalisation must be greater than zero, while trading volume cannot be negative. Values that fail these validity conditions are treated as missing and are removed if they affect variables required for the analysis.

Duplicate calendar dates are removed because the analytical design requires one market observation per day. Finally, the dataset is restricted to the common study period of **15 September 2025 to 23 May 2026**. The original `market_raw` DataFrame is not modified.

```python
# Inspect the raw market data
print(
    "Raw market shape:",
    market_raw.shape
)

print(
    "\nRaw data types:"
)

print(
    market_raw.dtypes
)

print(
    "\nMissing values:"
)

print(
    market_raw.isna().sum()
)

print(
    "\nExact duplicate rows:",
    market_raw.duplicated().sum()
)

market_raw.head()

```

```python
# Clean the market dataset
def clean_market_data(
    df: pd.DataFrame,
    start_date: str,
    end_date: str
) -> pd.DataFrame:
    """
    Clean XRP daily market data and restrict it
    to the common study period.
    """

    required = {
        "date",
        "price_usd",
        "market_cap_usd",
        "market_volume_usd"
    }

    missing_columns = (
        required - set(df.columns)
    )

    if missing_columns:
        raise ValueError(
            f"Market data missing columns: "
            f"{sorted(missing_columns)}"
        )

    clean = df.copy()

    # Convert date
    clean["date"] = pd.to_datetime(
        clean["date"],
        errors="coerce",
        utc=True
    )

    # Convert financial variables
    numeric_columns = [
        "price_usd",
        "market_cap_usd",
        "market_volume_usd"
    ]

    for column in numeric_columns:

        clean[column] = pd.to_numeric(
            clean[column],
            errors="coerce"
        )

    # Apply validity rules
    clean.loc[
        clean["price_usd"] <= 0,
        "price_usd"
    ] = np.nan

    clean.loc[
        clean["market_cap_usd"] <= 0,
        "market_cap_usd"
    ] = np.nan

    clean.loc[
        clean["market_volume_usd"] < 0,
        "market_volume_usd"
    ] = np.nan

    # Remove unusable observations
    clean = clean.dropna(
        subset=[
            "date",
            "price_usd",
            "market_cap_usd",
            "market_volume_usd"
        ]
    )

    # One observation per calendar day
    clean["date"] = (
        clean["date"]
        .dt.normalize()
    )

    clean = clean.drop_duplicates(
        subset="date",
        keep="last"
    )

    # Restrict to common study period
    start = pd.Timestamp(
        start_date,
        tz="UTC"
    )

    end = pd.Timestamp(
        end_date,
        tz="UTC"
    )

    clean = clean[
        clean["date"].between(
            start,
            end
        )
    ]

    # Remove timezone after alignment
    clean["date"] = (
        clean["date"]
        .dt.tz_localize(None)
    )

    clean = (
        clean
        .sort_values("date")
        .reset_index(drop=True)
    )

    return clean
```

## 4.2 XRP Ledger Data

The raw XRP Ledger dataset contains daily network activity obtained from the Coin Metrics public archive. The principal variable is `TxCnt`, which represents the number of transactions recorded on the XRP Ledger during each daily interval. Before this variable is used in the statistical analysis, the dataset is checked for invalid timestamps, missing or negative transaction counts, duplicate dates and observations outside the common study period.

The timestamp variable is converted to a standard datetime representation so that the network observations can later be aligned accurately with the market and media datasets. Transaction counts are converted to numeric values and records that cannot be interpreted reliably are treated as invalid. Negative transaction counts are not meaningful and are therefore excluded, while a value of zero is retained because it represents a theoretically valid daily observation.

Cleaning is performed on a copy of the raw data. The original `network_raw` DataFrame remains unchanged, while the processed observations are stored separately as `network_clean`.

```python
#Inspect the raw network data
print(
    "Raw network shape:",
    network_raw.shape
)

print(
    "\nRaw data types:"
)

print(
    network_raw.dtypes
)

print(
    "\nMissing values:"
)

print(
    network_raw.isna().sum()
)

print(
    "\nExact duplicate rows:",
    network_raw.duplicated().sum()
)

network_raw.head()
```

```python
#Identify invalid records before cleaning
network_dates_check = pd.to_datetime(
    network_raw["time"],
    errors="coerce",
    utc=True
)

network_counts_check = pd.to_numeric(
    network_raw["TxCnt"],
    errors="coerce"
)


invalid_network_records = pd.Series({
    "invalid_timestamps":
        int(
            network_dates_check
            .isna()
            .sum()
        ),

    "missing_or_invalid_transaction_counts":
        int(
            network_counts_check
            .isna()
            .sum()
        ),

    "negative_transaction_counts":
        int(
            (
                network_counts_check < 0
            ).sum()
        ),

    "duplicate_dates":
        int(
            network_dates_check
            .duplicated()
            .sum()
        )
})


invalid_network_records
```

The preliminary validation separates invalid records from valid zero-transaction observations. A missing or non-numeric transaction count cannot be used in the analysis, whereas a recorded count of zero remains a legitimate value. This distinction avoids unnecessarily altering the observed network series.

```python
#Clean the XRPL Data

def clean_network_data(
    df: pd.DataFrame,
    start_date: str,
    end_date: str
) -> pd.DataFrame:
    """
    Clean daily XRP Ledger transaction data.
    """

    required = {
        "time",
        "TxCnt"
    }

    missing_columns = (
        required
        - set(df.columns)
    )

    if missing_columns:

        raise ValueError(
            f"Network data missing columns: "
            f"{sorted(missing_columns)}"
        )

    clean = df.copy()

    # Process timestamps
    clean["time"] = pd.to_datetime(
        clean["time"],
        errors="coerce",
        utc=True
    )

    # Process transaction counts
    clean["TxCnt"] = pd.to_numeric(
        clean["TxCnt"],
        errors="coerce"
    )

    # Remove unusable records
    clean = clean.dropna(
        subset=[
            "time",
            "TxCnt"
        ]
    )

    # Transaction counts cannot be negative
    clean = clean[
        clean["TxCnt"] >= 0
    ]

    # Convert timestamps to daily dates
    clean["date"] = (
        clean["time"]
        .dt.normalize()
    )

    # Keep one observation per day
    clean = clean.drop_duplicates(
        subset="date",
        keep="last"
    )

    # Restrict to the common study period
    start = pd.Timestamp(
        start_date,
        tz="UTC"
    )

    end = pd.Timestamp(
        end_date,
        tz="UTC"
    )

    clean = clean[
        clean["date"].between(
            start,
            end
        )
    ]

    # Remove timezone for later merging
    clean["date"] = (
        clean["date"]
        .dt.tz_localize(None)
    )

    clean = clean.rename(
        columns={
            "TxCnt":
                "transaction_count"
        }
    )

    clean = (
        clean[
            [
                "date",
                "transaction_count"
            ]
        ]
        .sort_values("date")
        .reset_index(drop=True)
    )

    return clean
```

No transaction counts are automatically interpolated because doing so would introduce artificial blockchain activity into the dataset and could influence subsequent correlations with market volatility.

## 4.3 Media Data

The raw media dataset contains XRP-related article metadata obtained from the public news archive. Before sentiment analysis is performed, the observations are checked for duplicate records, missing or unusable headlines, invalid publication timestamps and malformed URLs. Cleaning is performed on a copy of the raw dataset so that the original media observations remain unchanged.

Duplicate handling requires particular care because financial news may be syndicated across several publishers. Exact duplicate records and repeated article identifiers are removed, while repeated headlines published on the same calendar day are treated as potential syndicated duplicates. This reduces the risk that the same news story is counted several times when daily sentiment and media-attention measures are calculated.

Headline text is cleaned conservatively using regular expressions. URLs embedded within headline text are removed and repeated whitespace is normalised, but punctuation and capitalisation are retained because these features may contribute to the VADER sentiment score. URL fields are validated separately. Articles with invalid URLs are flagged rather than automatically discarded because the URL itself is not required for headline-based sentiment analysis.

URL validity is assessed independently from headline validity. An article may still contribute useful textual information even when its source URL is missing or malformed. The cleaning procedure therefore records URL validity as a separate variable rather than treating an invalid URL as sufficient reason to discard the observation.

```python
#MEdia dataset cleaning
import re


def clean_news_data(
    df: pd.DataFrame
) -> pd.DataFrame:
    """
    Clean XRP-related media observations.
    """

    required = {
        "document_id",
        "published_at",
        "headline",
        "source_name",
        "url"
    }

    missing_columns = (
        required
        - set(df.columns)
    )

    if missing_columns:

        raise ValueError(
            f"Media data missing columns: "
            f"{sorted(missing_columns)}"
        )

    clean = df.copy()

    # Process timestamps
    clean["published_at"] = pd.to_datetime(
        clean["published_at"],
        errors="coerce",
        utc=True
    )

    # Standardise headline text
    clean["headline"] = (
        clean["headline"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    # Basic regex cleaning
    clean["clean_headline"] = (
        clean["headline"]
        .str.replace(
            r"https?://\S+",
            "",
            regex=True
        )
        .str.replace(
            r"\s+",
            " ",
            regex=True
        )
        .str.strip()
    )

    # Standardise source and URL text
    clean["source_name"] = (
        clean["source_name"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    clean["url"] = (
        clean["url"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    # Flag URL validity
    clean["valid_url"] = (
        clean["url"]
        .str.match(
            r"^https?://",
            case=False,
            na=False
        )
    )

    # Remove records without usable date or headline
    clean = clean[
        clean["published_at"].notna()
        & clean["clean_headline"].ne("")
    ]

    # Remove repeated article identifiers
    clean = clean.drop_duplicates(
        subset="document_id",
        keep="first"
    )

    # Create calendar date
    clean["date"] = (
        clean["published_at"]
        .dt.normalize()
        .dt.tz_localize(None)
    )

    # Remove likely syndicated duplicates
    clean = clean.drop_duplicates(
        subset=[
            "date",
            "clean_headline"
        ],
        keep="first"
    )

    clean = (
        clean
        .rename(
            columns={
                "source_name": "source"
            }
        )
        .sort_values("published_at")
        .reset_index(drop=True)
    )

    return clean
```

## 4.5 Unit Testing of the Headline-Cleaning Function

A focused test-driven development exercise is used to verify the headline-cleaning procedure. The purpose is not to test every individual Pandas operation, but to demonstrate how expected behaviour can be specified before a reusable function is accepted into the analytical pipeline. Headline preprocessing was selected because inappropriate text cleaning could directly affect the subsequent sentiment analysis. The required behaviour is therefore defined before the final helper function is implemented. The function must remove embedded URLs, normalise repeated whitespace, preserve punctuation and capitalisation, and return an empty string when supplied with missing text. The tests are written first, followed by the implementation and execution of the test suite. This provides a concise example of test-driven development while keeping testing focused on behaviour that could materially influence the analysis.

```python
#Define the tests first
import unittest


class TestCleanHeadlineText(
    unittest.TestCase
):

    def test_removes_url(self):

        result = clean_headline_text(
            "XRP rises https://example.com"
        )

        self.assertEqual(
            result,
            "XRP rises"
        )


    def test_removes_extra_spaces(self):

        result = clean_headline_text(
            "XRP   market   rises"
        )

        self.assertEqual(
            result,
            "XRP market rises"
        )


    def test_preserves_case_and_punctuation(self):

        result = clean_headline_text(
            "XRP SURGES!"
        )

        self.assertEqual(
            result,
            "XRP SURGES!"
        )


    def test_handles_missing_text(self):

        result = clean_headline_text(
            None
        )

        self.assertEqual(
            result,
            ""
        )
```

```python
#implement the function
def clean_headline_text(
    text
) -> str:
    """
    Apply minimal cleaning suitable for VADER sentiment analysis.
    """

    if pd.isna(text):
        return ""

    text = str(text)

    # Remove embedded URLs
    text = re.sub(
        r"https?://\S+",
        "",
        text
    )

    # Normalise repeated whitespace
    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()
```

```python
#run the unit test
test_suite = (
    unittest
    .defaultTestLoader
    .loadTestsFromTestCase(
        TestCleanHeadlineText
    )
)


test_runner = unittest.TextTestRunner(
    verbosity=2
)


test_result = test_runner.run(
    test_suite
)


assert test_result.wasSuccessful()


print(
    "\nAll headline-cleaning tests passed."
)
```

```python
#Apply the tested function to the cleaned media data
news_clean[
    "clean_headline"
] = (
    news_clean[
        "headline"
    ]
    .apply(
        clean_headline_text
    )
)


assert news_clean[
    "clean_headline"
].ne("").all()


print(
    "Tested headline-cleaning function "
    "applied successfully."
)
```

### TDD Evaluation

The four tests confirm that the headline-cleaning function satisfies the behaviour specified before implementation. URLs and unnecessary whitespace are removed, while capitalisation and punctuation are deliberately preserved because these features may contribute to VADER's interpretation of sentiment intensity. Missing input is handled safely by returning an empty string. 
This test-driven example demonstrates how a small but analytically important transformation can be validated independently before being applied to the full dataset. Testing is concentrated on behaviour that could affect downstream results rather than attempting to create unit tests for every simple DataFrame operation. Additional assertions throughout the notebook provide complementary checks on dataset structure, missing values, duplicate dates and numerical validity.

## 5.1 Market Feature Engineering

The cleaned XRP market dataset contains daily price, market capitalisation and trading volume. These raw variables are transformed into measures of market movement, volatility and trading activity that are required for the hypotheses and the later exploratory index. Daily return is calculated as the percentage change in XRP price between consecutive observations. Absolute return represents the magnitude of the daily price movement regardless of direction and is used in the first hypothesis. Rolling volatility is calculated as the standard deviation of daily returns over seven-day and thirty-day windows. The seven-day measure is used as the principal short-term volatility variable, while the thirty-day measure provides a longer-term comparison. Trading volume is transformed using the natural logarithm because cryptocurrency volume can contain very large observations and strong positive skew. A standardised volume score is also calculated so that unusually high or low trading activity can later be compared with variables measured on different numerical scales. All transformations are performed on a copy of the cleaned market dataset.

```python
#standadisation helper
def standardise_series(
    series: pd.Series
) -> pd.Series:
    """
    Convert a numeric series to z-scores.
    """

    numeric = pd.to_numeric(
        series,
        errors="coerce"
    )

    mean = numeric.mean()
    std = numeric.std(ddof=0)

    if pd.isna(std) or std == 0:
        return pd.Series(
            np.nan,
            index=series.index
        )

    return (
        numeric - mean
    ) / std
```

```python
#Create mkt features
def add_market_features(
    df: pd.DataFrame
) -> pd.DataFrame:
    """
    Create XRP market variables used in the analysis.
    """

    result = (
        df.copy()
        .sort_values("date")
        .reset_index(drop=True)
    )

    # Daily percentage return
    result["daily_return"] = (
        result["price_usd"]
        .pct_change(
            fill_method=None
        )
    )

    # Magnitude of daily market movement
    result["absolute_return"] = (
        result["daily_return"]
        .abs()
    )

    # Rolling volatility
    result["rolling_volatility_7d"] = (
        result["daily_return"]
        .rolling(
            window=7,
            min_periods=7
        )
        .std()
    )

    result["rolling_volatility_30d"] = (
        result["daily_return"]
        .rolling(
            window=30,
            min_periods=30
        )
        .std()
    )

    # Daily change in trading volume
    result["volume_change"] = (
        result["market_volume_usd"]
        .pct_change(
            fill_method=None
        )
    )

    # Reduce positive skew in trading volume
    result["log_market_volume"] = (
        np.log1p(
            result["market_volume_usd"]
        )
    )

    # Standardised variables
    result["volume_zscore"] = (
        standardise_series(
            result["log_market_volume"]
        )
    )

    result["absolute_return_zscore"] = (
        standardise_series(
            result["absolute_return"]
        )
    )

    return result
```

## 5.2 Sentiment Analysis

Headline sentiment is estimated using VADER, a pre-trained rule-based sentiment model designed for short textual content. VADER is appropriate for this project because the media dataset contains short news headlines and no labelled XRP-specific training dataset is available for developing a supervised sentiment model. The analysis uses the VADER compound score, which ranges from negative to positive sentiment. Headlines with a compound score of at least 0.05 are classified as positive, scores of -0.05 or below are classified as negative, and values between these thresholds are classified as neutral. Sentiment is first calculated for each cleaned headline and is then aggregated to the daily level so that it can be aligned with the market and network datasets. The principal daily variable is mean headline sentiment, while article count is retained as a measure of media attention. Median sentiment and the proportions of positive and negative headlines are retained as supporting descriptive measures.

```python
#Load CADER
from vaderSentiment.vaderSentiment import (
    SentimentIntensityAnalyzer
)


sentiment_analyser = (
    SentimentIntensityAnalyzer()
)


print(
    "VADER sentiment model loaded."
)
```

```python
#Score the headlines
def score_sentiment(
    text: str
) -> float:
    """
    Return the VADER compound sentiment score.
    """

    return sentiment_analyser.polarity_scores(
        text
    )["compound"]
```

```python
#Create one sentiment score for every cleaned headline
news_sentiment = news_clean.copy()


news_sentiment["sentiment_score"] = (
    news_sentiment[
        "clean_headline"
    ]
    .apply(
        score_sentiment
    )
)


news_sentiment.head()
```

```python
#create sentiment categories
news_sentiment[
    "sentiment_label"
] = np.select(
    [
        news_sentiment[
            "sentiment_score"
        ] >= 0.05,

        news_sentiment[
            "sentiment_score"
        ] <= -0.05
    ],

    [
        "positive",
        "negative"
    ],

    default="neutral"
)
```

```python
#Validate the sentiment scores
print(
    "Missing sentiment scores:",
    news_sentiment[
        "sentiment_score"
    ].isna().sum()
)


print(
    "Minimum score:",
    news_sentiment[
        "sentiment_score"
    ].min()
)


print(
    "Maximum score:",
    news_sentiment[
        "sentiment_score"
    ].max()
)


print(
    "\nSentiment categories:"
)

print(
    news_sentiment[
        "sentiment_label"
    ].value_counts()
)
```

```python
assert news_sentiment[
    "sentiment_score"
].between(
    -1,
    1
).all()


assert news_sentiment[
    "sentiment_score"
].notna().all()


print(
    "Sentiment validation passed."
)
```

The article-level validation confirms that every retained headline received a sentiment score within VADER's expected range. The categorical labels are used primarily for descriptive interpretation, while the continuous compound score is retained for the statistical analysis because it preserves more information about sentiment intensity.

```python
#Aggragate sentiment by day
sentiment_daily = (
    news_sentiment
    .groupby(
        "date",
        as_index=False
    )
    .agg(
        mean_sentiment=(
            "sentiment_score",
            "mean"
        ),

        median_sentiment=(
            "sentiment_score",
            "median"
        ),

        article_count=(
            "sentiment_score",
            "size"
        ),

        positive_ratio=(
            "sentiment_label",
            lambda x:
            (x == "positive").mean()
        ),

        negative_ratio=(
            "sentiment_label",
            lambda x:
            (x == "negative").mean()
        )
    )
)


sentiment_daily.head()
```

## 5.3 Network Aggregation

The cleaned XRP Ledger dataset already contains one transaction-count observation for each calendar day, so no additional temporal aggregation is required. Further aggregation would unnecessarily reduce the temporal resolution of the network data. The daily transaction count is retained as the principal network-activity variable for hypothesis testing. A logarithmic transformation is also created to reduce the influence of unusually large transaction counts, while a standardised network-activity score is calculated for later use in the exploratory composite index. A seven-day moving average is retained for descriptive analysis of broader network-activity trends. The original `network_clean` DataFrame remains unchanged. The transformed observations are stored separately as `network_features`.


```python
#Creating network features
def add_network_features(
    df: pd.DataFrame
) -> pd.DataFrame:
    """
    Create daily XRP Ledger network features.
    """

    result = (
        df.copy()
        .sort_values("date")
        .reset_index(drop=True)
    )

    # Daily percentage change
    result["network_activity_change"] = (
        result["transaction_count"]
        .pct_change(
            fill_method=None
        )
    )

    # Log transformation
    result["log_transaction_count"] = (
        np.log1p(
            result["transaction_count"]
        )
    )

    # Standardised network activity
    result["network_activity_zscore"] = (
        standardise_series(
            result["log_transaction_count"]
        )
    )

    # Descriptive seven-day moving average
    result["network_activity_7d_ma"] = (
        result["transaction_count"]
        .rolling(
            window=7,
            min_periods=1
        )
        .mean()
    )

    return result
```

```python
#apply the transformation
network_features = add_network_features(
    network_clean
)


print(
    "Network feature dataset shape:",
    network_features.shape
)


network_features.head()
```

```python
#Validating the features
assert (
    network_features[
        "transaction_count"
    ] >= 0
).all()


assert np.isfinite(
    network_features[
        "log_transaction_count"
    ]
).all()


assert network_features[
    "date"
].is_unique


print(
    "Network feature validation passed."
)
```

The raw transaction count is retained for the main network hypothesis because it has a direct interpretation as daily XRP Ledger activity. The logarithmic and standardised measures are supplementary transformations intended for descriptive analysis and construction of variables that require a common numerical scale. The first percentage-change observation is expected to be missing because no previous day is available for comparison.

## 5.4 SQL Aggregation Demonstration

A small SQLite workflow is used to demonstrate relational storage and aggregation of the article-level sentiment observations. SQL is appropriate for operations in which records are grouped according to a common key and summarised using aggregate functions. The cleaned article-level sentiment data are stored temporarily in an in-memory SQLite database. A SQL `GROUP BY` query then calculates daily mean sentiment and article count. These results are compared with the equivalent Pandas aggregation used elsewhere in the project. The purpose is to demonstrate an alternative data-processing method without unnecessarily replacing Pandas for the time-series transformations required later in the analysis.

```python
#Temporary SQL table
import sqlite3


sql_news = news_sentiment[
    [
        "date",
        "sentiment_score"
    ]
].copy()


sql_news["date"] = (
    pd.to_datetime(
        sql_news["date"]
    )
    .dt.strftime("%Y-%m-%d")
)


connection = sqlite3.connect(
    ":memory:"
)


sql_news.to_sql(
    "news_sentiment",
    connection,
    index=False,
    if_exists="replace"
)


print(
    "Article-level sentiment data stored in SQLite."
)
```

```python
#SQL aggregate
sql_query = """
SELECT
    date,
    COUNT(*) AS article_count,
    AVG(sentiment_score) AS mean_sentiment
FROM news_sentiment
GROUP BY date
ORDER BY date
"""


sql_sentiment_daily = pd.read_sql_query(
    sql_query,
    connection
)


sql_sentiment_daily["date"] = (
    pd.to_datetime(
        sql_sentiment_daily["date"]
    )
)


sql_sentiment_daily.head()
```

```python
#Compare SQL with Pandas
comparison = sentiment_daily[
    [
        "date",
        "article_count",
        "mean_sentiment"
    ]
].merge(
    sql_sentiment_daily,
    on="date",
    suffixes=(
        "_pandas",
        "_sql"
    )
)


article_count_match = (
    comparison[
        "article_count_pandas"
    ]
    .eq(
        comparison[
            "article_count_sql"
        ]
    )
    .all()
)


sentiment_match = np.allclose(
    comparison[
        "mean_sentiment_pandas"
    ],
    comparison[
        "mean_sentiment_sql"
    ],
    equal_nan=True
)


print(
    "Article counts match:",
    article_count_match
)

print(
    "Mean sentiment values match:",
    sentiment_match
)
```

```python
#Close the database
connection.close()
```

The SQL and Pandas results are expected to agree because both procedures perform the same grouping operation. SQL provides a concise approach to relational grouping and aggregation, whereas Pandas is retained for the main analytical pipeline because it integrates more naturally with rolling windows, time shifts, statistical functions and subsequent visualisation. Demonstrating both approaches therefore illustrates tool selection according to the structure of the task rather than using one library for every operation.

## 6.2 Statistical Method Selection

The choice of statistical method was based on the distributional characteristics of the variables used in each hypothesis. Pearson correlation assumes that the relationship between two continuous variables is approximately linear and is sensitive to strong skewness and influential outliers. Spearman rank correlation does not require normally distributed variables and is more appropriate when relationships are monotonic but the underlying distributions are skewed or contain extreme observations. The exploratory analysis identified several characteristics that may violate the assumptions of Pearson correlation. Trading volume contains substantial positive skew and several large spikes, while absolute returns and rolling volatility also contain extreme observations. XRP Ledger transaction activity includes periods of unusually high network activity, and the media-sentiment dataset contains fewer observations and irregular temporal coverage. Distributional diagnostics are therefore examined using skewness and the Shapiro-Wilk test. These diagnostics are interpreted alongside the visual evidence from the exploratory analysis rather than being used as the sole basis for selecting a statistical method.

```python
#Distribution diagnostics
def distribution_summary(
    series: pd.Series,
    name: str
) -> dict:
    """
    Return basic distribution diagnostics.
    """

    data = pd.to_numeric(
        series,
        errors="coerce"
    ).dropna()

    shapiro_stat, shapiro_p = (
        stats.shapiro(data)
    )

    return {
        "variable": name,
        "n": len(data),
        "skewness": data.skew(),
        "shapiro_statistic": shapiro_stat,
        "shapiro_p_value": shapiro_p
    }
```

```python
# Apply diagnostics to hypothesis variables

diagnostic_variables = {
    "Trading volume":
        analysis_final["market_volume_usd"],

    "Absolute return":
        analysis_final["absolute_return"],

    "Transaction count":
        analysis_final["transaction_count"],

    "7-day volatility":
        analysis_final["rolling_volatility_7d"],

    "Mean sentiment":
        analysis_final["mean_sentiment"],

    "Next-day return":
        analysis_final["next_day_return"]
}


distribution_results = pd.DataFrame(
    [
        distribution_summary(
            series,
            name
        )
        for name, series
        in diagnostic_variables.items()
    ]
)


distribution_results[
    [
        "variable",
        "n",
        "skewness",
        "shapiro_p_value"
    ]
].round(4)
```

### Distribution Diagnostics Interpretation

The distribution diagnostics provide strong evidence that the principal hypothesis variables are not normally distributed. Trading volume, absolute return, transaction count and seven-day volatility all display substantial positive skewness, while their Shapiro-Wilk p-values are below 0.05. This indicates that the normality assumption required for Pearson correlation is not satisfied. Mean sentiment is less strongly skewed, but its Shapiro-Wilk p-value is also below 0.05, indicating a departure from normality. Next-day return shows only moderate positive skewness, yet it likewise fails the normality test. Taken together with the extreme observations identified during exploratory analysis, these results support the use of **Spearman rank correlation** as the primary inferential method for H1, H2 and H3. Spearman correlation is more appropriate because it does not require normally distributed variables and is less sensitive to extreme observations than Pearson correlation.

The histograms confirm the distributional patterns identified by the numerical diagnostics. Trading volume, absolute return, transaction count and seven-day volatility are visibly right-skewed, with most observations concentrated at lower values and a smaller number of extreme observations extending the upper tails. Mean sentiment is more concentrated around neutral values but remains irregular, while next-day returns are centred close to zero with several unusually large positive and negative observations. These visual patterns, together with the significant Shapiro-Wilk results, indicate that the assumptions required for Pearson correlation are not well supported. Spearman rank correlation is therefore selected for the three main hypothesis tests because it is less sensitive to non-normal distributions and extreme observations while remaining appropriate for assessing monotonic associations between the variables.

### Statistical Assumptions

Spearman correlation requires paired observations that can be ranked and assumes that the relationship is broadly monotonic rather than necessarily linear. It is less sensitive than Pearson correlation to non-normal distributions and extreme observations, although it does not eliminate the influence of all unusual data points. The observations in this project are also ordered through time and may therefore exhibit temporal dependence. The standard correlation tests do not fully model serial correlation, so the results are interpreted as evidence of association rather than causal relationships. This limitation is considered again in the final evaluation.

### Method Selection

Based on the exploratory plots and distributional diagnostics, **Spearman rank correlation is selected as the primary inferential method for all three hypotheses**. This provides a consistent approach across variables that display skewness, extreme observations and unequal sample sizes. For H1, Spearman correlation tests the association between daily XRP trading volume and absolute daily return. For H2, it tests the association between XRP Ledger transaction activity and seven-day rolling volatility. For H3, it tests the association between daily headline sentiment and the following day's XRP return. For each hypothesis, the analysis reports the number of complete paired observations, Spearman's correlation coefficient, the p-value and a bootstrap 95% confidence interval. Statistical significance is assessed using \(\alpha = 0.05\). The magnitude and direction of the coefficient are interpreted alongside statistical significance because a small p-value alone does not indicate that an association is practically important.

Overall, Spearman correlation provides the most appropriate balance between statistical robustness and interpretability for the present dataset. The method addresses the non-normal and outlier-prone characteristics observed in the market and network variables while remaining suitable for the smaller and more irregular sentiment sample. The same primary method is therefore used across the three hypotheses, allowing the results to be compared consistently.

## 6.3 Hypothesis 1: Trading Volume and Absolute Returns

The first hypothesis examines whether higher XRP trading activity is associated with larger daily price movements. The null hypothesis states that there is no significant association between daily XRP trading volume and absolute daily return. The alternative hypothesis proposes a positive association, whereby higher trading volume tends to occur on days with larger price movements. Spearman rank correlation is used because both trading volume and absolute return are positively skewed and contain extreme observations. The test is conducted using complete paired observations only, with statistical significance evaluated at the 5% level.

```python
def spearman_with_ci(
    x,
    y,
    n_bootstrap=2000,
    seed=42
):
    """
    Calculate Spearman correlation, p-value and
    percentile bootstrap 95% confidence interval.
    """

    paired = pd.concat(
        [x, y],
        axis=1
    ).dropna()

    paired.columns = [
        "x",
        "y"
    ]

    rho, p_value = stats.spearmanr(
        paired["x"],
        paired["y"]
    )

    rng = np.random.default_rng(
        seed
    )

    bootstrap_rho = []

    for _ in range(n_bootstrap):

        sample = paired.iloc[
            rng.integers(
                0,
                len(paired),
                len(paired)
            )
        ]

        if (
            sample["x"].nunique() < 2
            or sample["y"].nunique() < 2
        ):
            continue

        value = stats.spearmanr(
            sample["x"],
            sample["y"]
        ).statistic

        bootstrap_rho.append(
            value
        )

    ci_low, ci_high = np.percentile(
        bootstrap_rho,
        [2.5, 97.5]
    )

    return {
        "n": len(paired),
        "spearman_rho": rho,
        "p_value": p_value,
        "ci_95_low": ci_low,
        "ci_95_high": ci_high
    }
```

```python
#H1 spearman's correlation

h1_result = spearman_with_ci(
    analysis_final[
        "market_volume_usd"
    ],
    analysis_final[
        "absolute_return"
    ]
)


h1_table = pd.DataFrame(
    [h1_result],
    index=["H1"]
).round(4)


h1_table
```

### H1 Result Interpretation

The first hypothesis produced a moderate positive Spearman correlation between XRP trading volume and absolute daily return, with \(\rho = 0.4566\) based on 350 paired observations. The p-value is below 0.05, and the 95% bootstrap confidence interval ranges from 0.3658 to 0.5448, remaining entirely above zero. These results provide evidence against the null hypothesis and support the alternative hypothesis that higher XRP trading volume is associated with larger daily price movements. The relationship is statistically significant and moderate in strength, although it should still be interpreted as an association rather than evidence that increased trading volume causes larger returns.

```python
#H1 scatter plot
plt.figure(
    figsize=(7, 5)
)

plt.scatter(
    analysis_final[
        "log_market_volume"
    ],
    analysis_final[
        "absolute_return"
    ],
    alpha=0.6
)

plt.title(
    "XRP Trading Volume and Absolute Daily Return"
)

plt.xlabel(
    "Log Trading Volume"
)

plt.ylabel(
    "Absolute Daily Return"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()
```

The scatter plot shows a clear positive monotonic pattern between log trading volume and absolute daily XRP return. Most observations are concentrated at lower return levels, but larger absolute returns become more common as trading volume increases. Several extreme observations are also visible at high-volume levels, which supports the earlier decision to use Spearman rank correlation rather than a method that depends strongly on linearity and normality. The visual pattern is consistent with the formal H1 result of \(\rho = 0.4566\), indicating a moderate positive association. ## 6.6 Time-Lag Analysis

Time-lag analysis is a statistical and data-driven method used to measure the time delay or temporal relationship between an event, variable, or interaction and its resulting effect. The third hypothesis examines only the predefined one-day relationship between media sentiment and subsequent XRP return. An additional exploratory time-lag analysis is conducted to investigate whether the association changes when returns before and after the sentiment observation are considered. Lags from -7 to +7 calendar days are examined. A positive lag indicates that sentiment is compared with a future XRP return; for example, lag +1 compares today's sentiment with tomorrow's return. A negative lag compares today's sentiment with an earlier market return and can therefore indicate whether media tone may be responding to preceding market movements. Because fifteen lag relationships are examined, the analysis increases the risk of obtaining an apparently significant result by chance. Bonferroni-adjusted p-values are therefore reported alongside the unadjusted values. The lag analysis remains exploratory and does not replace the predefined H3 test.

```python
#Calculating lagged correlations
lag_results = []


for lag in range(
    -7,
    8
):

    future_or_past_return = (
        analysis_final[
            "daily_return"
        ]
        .shift(-lag)
    )

    paired = pd.concat(
        [
            analysis_final[
                "mean_sentiment"
            ],
            future_or_past_return
        ],
        axis=1
    ).dropna()

    paired.columns = [
        "sentiment",
        "return"
    ]

    if len(paired) >= 3:

        rho, p_value = stats.spearmanr(
            paired[
                "sentiment"
            ],
            paired[
                "return"
            ]
        )

        lag_results.append({
            "lag_days": lag,
            "n": len(paired),
            "spearman_rho": rho,
            "p_value": p_value
        })


lag_results = pd.DataFrame(
    lag_results
)
```

```python
#Correct for multiple comparisons
number_of_tests = len(
    lag_results
)


lag_results[
    "bonferroni_p"
] = np.minimum(
    lag_results[
        "p_value"
    ]
    * number_of_tests,
    1.0
)


lag_results[
    "significant_after_correction"
] = (
    lag_results[
        "bonferroni_p"
    ] < 0.05
)


lag_results.round(4)
```

```python
#Plot lagged correlations
plt.figure(
    figsize=(9, 5)
)

plt.plot(
    lag_results[
        "lag_days"
    ],
    lag_results[
        "spearman_rho"
    ],
    marker="o"
)

plt.axhline(
    0,
    linewidth=1
)

plt.axvline(
    0,
    linestyle="--",
    linewidth=1
)

plt.xticks(
    range(-7, 8)
)

plt.title(
    "Lagged Relationship Between XRP Sentiment and Returns"
)

plt.xlabel(
    "Lag (Days)"
)

plt.ylabel(
    "Spearman Correlation"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()
```

### Time-Lag Analysis Interpretation

The lag analysis shows that the association between XRP headline sentiment and returns varies across the \(-7\) to \(+7\)-day window. The strongest positive correlations appear around lags of -2, -1 and +1 days, while several other lags are close to zero or slightly negative. The positive correlation at lag +1 is consistent with the direction observed in H3, but the overall lag pattern is irregular rather than showing a stable predictive relationship. The relatively strong negative-lag correlations also suggest that media sentiment may partly reflect market movements that have already occurred rather than consistently preceding future returns. These results should therefore be treated as exploratory. The Bonferroni-adjusted p-values provide the appropriate basis for determining whether any individual lag remains statistically significant after accounting for the multiple comparisons.

```python
#Identifying the strongest observed lag
strongest_lag = (
    lag_results
    .loc[
        lag_results[
            "spearman_rho"
        ]
        .abs()
        .idxmax()
    ]
)


strongest_lag.to_frame(
    name="result"
)
```

### Strongest Lag Interpretation

The strongest observed lag occurs at **-1 day**, with a Spearman correlation of \(\rho = 0.2222\) based on 63 paired observations. This indicates a weak positive association between current headline sentiment and the previous day's XRP return, suggesting that media tone may partly reflect market movements that have already occurred. However, the relationship is not statistically significant (\(p = 0.0800\)), and the Bonferroni-adjusted p-value is 1.000. The result therefore does not remain significant after correcting for multiple lag comparisons. Overall, the lag analysis provides no statistically robust evidence that XRP headline sentiment consistently predicts future returns. The strongest observed relationship instead occurs at a negative lag, which is more consistent with sentiment responding to recent market behaviour than leading it.

## 6.7 XRP Market Attention and Stress Index

An exploratory **XRP Market Attention and Stress Index (XMASI)** is developed to combine information from the three independent data domains used in this study. The purpose of the index is to provide a single standardised measure representing periods in which XRP experiences unusually high market movement, trading activity, network activity, media attention and negative headline sentiment. Five components are included: absolute daily return, trading volume, XRP Ledger transaction activity, article count and headline sentiment. Absolute return represents market stress, while trading volume, transaction count and article frequency represent different forms of activity or attention. Sentiment enters the index with its sign reversed so that more negative headline sentiment contributes positively to the stress measure. Because the variables are measured on different numerical scales, each component is converted to a z-score before combination. Equal weighting is used to maintain transparency and to avoid estimating weights from the same dataset on which the index is evaluated. Higher XMASI values therefore indicate days characterised by relatively greater combined market stress and attention.

The index is defined as:

\[
XMASI_t =
\frac{
Z(|R_t|)
+
Z(\log(1+V_t))
+
Z(\log(1+N_t))
+
Z(\log(1+A_t))
+
Z(-S_t)
}{5}
\]

where \(R_t\) is daily XRP return, \(V_t\) is trading volume, \(N_t\) is XRP Ledger transaction count, \(A_t\) is XRP-related article count and \(S_t\) is mean daily headline sentiment.

The index is calculated only on dates for which all five components are observed. In particular, missing sentiment is not replaced with zero because a day without retained news coverage should not automatically be interpreted as a day of neutral sentiment.

```python
#Construct the components:

xmasi_data = analysis_final.copy()


# Market stress
xmasi_data[
    "return_stress_z"
] = standardise_series(
    xmasi_data[
        "absolute_return"
    ]
)


# Trading activity
xmasi_data[
    "volume_stress_z"
] = standardise_series(
    np.log1p(
        xmasi_data[
            "market_volume_usd"
        ]
    )
)


# Network activity
xmasi_data[
    "network_stress_z"
] = standardise_series(
    np.log1p(
        xmasi_data[
            "transaction_count"
        ]
    )
)


# Media attention
xmasi_data[
    "attention_stress_z"
] = standardise_series(
    np.log1p(
        xmasi_data[
            "article_count"
        ]
    )
)


# Negative sentiment contributes positively to stress
xmasi_data[
    "sentiment_stress_z"
] = standardise_series(
    -xmasi_data[
        "mean_sentiment"
    ]
)
```

```python
#XMASI Calculation
xmasi_components = [
    "return_stress_z",
    "volume_stress_z",
    "network_stress_z",
    "attention_stress_z",
    "sentiment_stress_z"
]


xmasi_data["XMASI"] = (
    xmasi_data[
        xmasi_components
    ]
    .mean(
        axis=1,
        skipna=False
    )
)


print(
    "XMASI observations:",
    xmasi_data[
        "XMASI"
    ].notna().sum()
)


xmasi_data[
    [
        "date",
        "XMASI"
    ]
].dropna().head()
```

### Initial XMASI Interpretation

The first calculated XMASI observations show that the composite index varies substantially across days, with both positive and negative values. For example, 6 January 2026 records a relatively high XMASI value of approximately 1.25, indicating that the combined measures of market movement, trading activity, network activity, media attention and negative sentiment were elevated relative to their typical levels on that date. By contrast, negative values on 9 and 12 January indicate comparatively lower combined stress and attention. These initial observations confirm that the index is responding to variation across its underlying components. However, individual values should not be interpreted as trading signals; the usefulness of XMASI is evaluated formally by examining its association with subsequent XRP volatility.

```python
#Validate the index
assert np.isfinite(
    xmasi_data[
        "XMASI"
    ].dropna()
).all()


assert (
    xmasi_data[
        "XMASI"
    ].notna().sum()
    > 0
)


print(
    "XMASI validation passed."
)
```

```python
#XMASI Plot
xmasi_plot = xmasi_data.dropna(
    subset=["XMASI"]
)


plt.figure(
    figsize=(10, 5)
)

plt.plot(
    xmasi_plot["date"],
    xmasi_plot["XMASI"],
    marker="o"
)

plt.axhline(
    0,
    linewidth=1
)

plt.title(
    "XRP Market Attention and Stress Index (XMASI)"
)

plt.xlabel(
    "Date"
)

plt.ylabel(
    "XMASI"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()
```

### XMASI Time-Series Interpretation

The XMASI series shows substantial variation across the dates for which all five index components are available. The highest observed value occurs in early January 2026, reaching approximately 1.25, indicating an unusually elevated combination of market movement, trading activity, network activity, media attention and negative sentiment relative to the study-period averages. Several negative XMASI observations are also visible, particularly during January and April 2026, indicating periods of comparatively lower combined stress and attention. Later observations are generally closer to zero, although a moderate positive peak occurs in April. The irregular spacing of the plotted observations reflects the limited availability of media sentiment rather than missing market or network data. Consequently, XMASI should be interpreted as an exploratory indicator for dates with complete multi-source information rather than as a continuous daily market index.

### XMASI Evaluation

To provide an external evaluation of the composite measure, XMASI is compared with XRP volatility over the following seven calendar days. Future volatility is calculated using returns from day \(t+1\) through day \(t+7\), ensuring that the evaluation period does not include the current-day return used within the index itself. Spearman correlation is used because both the composite index and future volatility may contain extreme observations and are not assumed to follow normal distributions. This evaluation remains exploratory and does not establish that XMASI predicts or causes future market volatility.

```python
#Calculate future seven-day volatility
future_returns = pd.concat(
    [
        xmasi_data[
            "daily_return"
        ].shift(-lag)
        for lag in range(1, 8)
    ],
    axis=1
)


# Require all seven future returns
complete_future_window = (
    future_returns
    .notna()
    .sum(axis=1)
    == 7
)


xmasi_data[
    "future_volatility_7d"
] = (
    future_returns
    .std(
        axis=1
    )
    .where(
        complete_future_window
    )
)
```

```python
#Evaluate the relationship
xmasi_result = spearman_with_ci(
    xmasi_data[
        "XMASI"
    ],
    xmasi_data[
        "future_volatility_7d"
    ]
)


xmasi_result_table = pd.DataFrame(
    [xmasi_result],
    index=["XMASI"]
).round(4)


xmasi_result_table
```

```python
xmasi_evaluation = (
    xmasi_data
    .dropna(
        subset=[
            "XMASI",
            "future_volatility_7d"
        ]
    )
    .copy()
)


high_threshold = (
    xmasi_evaluation[
        "XMASI"
    ]
    .quantile(0.75)
)


xmasi_evaluation[
    "XMASI_group"
] = np.where(
    xmasi_evaluation[
        "XMASI"
    ] >= high_threshold,
    "High XMASI",
    "Other observations"
)


xmasi_group_summary = (
    xmasi_evaluation
    .groupby(
        "XMASI_group"
    )[
        "future_volatility_7d"
    ]
    .agg(
        [
            "count",
            "mean",
            "median"
        ]
    )
)


xmasi_group_summary
```

### XMASI Evaluation

The evaluation provides little evidence that higher XMASI values are associated with greater subsequent XRP volatility. Across the 20 observations with complete index components and seven-day future volatility, the Spearman correlation is weakly negative (\(\rho = -0.1519\)) and statistically non-significant (\(p = 0.5227\)). The 95% bootstrap confidence interval ranges from -0.5776 to 0.3247 and crosses zero substantially, indicating considerable uncertainty around the estimated relationship. The descriptive group comparison leads to the same conclusion. The five observations classified as high XMASI were followed by a mean seven-day volatility of approximately 0.0223, compared with approximately 0.0244 for the remaining observations. Future volatility was therefore slightly lower, rather than higher, following high-index observations.

These findings do not support the use of XMASI as a predictor of subsequent volatility in the present sample. However, the evaluation is limited by the small number of complete observations, largely because XMASI requires sentiment data alongside market and network measures. The index should therefore be regarded as an exploratory multi-source descriptive indicator rather than a validated forecasting measure. This negative result is retained because evaluating the proposed index critically is more informative than assuming that a novel measure must demonstrate predictive value.

```python
#Calculate future seven-day volatility
future_returns = pd.concat(
    [
        xmasi_data[
            "daily_return"
        ].shift(-lag)
        for lag in range(1, 8)
    ],
    axis=1
)


# Require all seven future returns
complete_future_window = (
    future_returns
    .notna()
    .sum(axis=1)
    == 7
)


xmasi_data[
    "future_volatility_7d"
] = (
    future_returns
    .std(
        axis=1
    )
    .where(
        complete_future_window
    )
)
```

```python
xmasi_result = spearman_with_ci(
    xmasi_data[
        "XMASI"
    ],
    xmasi_data[
        "future_volatility_7d"
    ]
)


xmasi_result_table = pd.DataFrame(
    [xmasi_result],
    index=["XMASI"]
).round(4)


xmasi_result_table
```

```python
xmasi_evaluation = (
    xmasi_data
    .dropna(
        subset=[
            "XMASI",
            "future_volatility_7d"
        ]
    )
    .copy()
)


high_threshold = (
    xmasi_evaluation[
        "XMASI"
    ]
    .quantile(0.75)
)


xmasi_evaluation[
    "XMASI_group"
] = np.where(
    xmasi_evaluation[
        "XMASI"
    ] >= high_threshold,
    "High XMASI",
    "Other observations"
)


xmasi_group_summary = (
    xmasi_evaluation
    .groupby(
        "XMASI_group"
    )[
        "future_volatility_7d"
    ]
    .agg(
        [
            "count",
            "mean",
            "median"
        ]
    )
)


xmasi_group_summary
```

### XMASI Evaluation

The relationship between XMASI and subsequent seven-day XRP volatility is weak and statistically non-significant. The Spearman correlation is \(\rho = -0.1519\) based on 20 complete observations, with \(p = 0.5227\). The 95% bootstrap confidence interval ranges from -0.5776 to 0.3247 and includes zero, indicating substantial uncertainty around the estimated association. The descriptive comparison produces a similar result. High-XMASI observations were followed by a mean future volatility of approximately 0.0223, compared with approximately 0.0244 for the remaining observations. Median future volatility was also slightly lower for the high-XMASI group. These findings do not support XMASI as a predictor of higher subsequent XRP volatility in the present sample. However, the evaluation is limited by the small number of complete observations, because the index requires market, network and sentiment information to be available on the same date. XMASI is therefore retained as an exploratory multi-source indicator rather than a validated forecasting measure.

## 6.8 Sensitivity and Robustness Checks

Sensitivity analysis is used to assess whether the principal findings depend strongly on particular variable definitions. The purpose is not to search repeatedly for statistically significant results, but to determine whether the direction and approximate strength of the main relationships remain reasonably stable when defensible alternative specifications are used.

Three alternative specifications are examined. H1 is repeated using logarithmic trading volume, H2 is repeated using fourteen-day rather than seven-day rolling volatility, and H3 is repeated using median rather than mean daily headline sentiment. A final leave-one-component-out analysis examines whether the exploratory XMASI evaluation is dominated by any single component.

```python
#Create 14-day volatility
robustness_data = analysis_final.copy()


robustness_data[
    "rolling_volatility_14d"
] = (
    robustness_data[
        "daily_return"
    ]
    .rolling(
        window=14,
        min_periods=14
    )
    .std()
)
```

```python
#Run the three alternative specifications
h1_log = spearman_with_ci(
    analysis_final[
        "log_market_volume"
    ],
    analysis_final[
        "absolute_return"
    ]
)


h2_14d = spearman_with_ci(
    robustness_data[
        "transaction_count"
    ],
    robustness_data[
        "rolling_volatility_14d"
    ]
)


h3_median = spearman_with_ci(
    analysis_final[
        "median_sentiment"
    ],
    analysis_final[
        "next_day_return"
    ]
)
```

```python
#Compare the original and alternative results
robustness_results = pd.DataFrame([
    {
        "analysis":
            "H1 original volume",
        **h1_result
    },

    {
        "analysis":
            "H1 log volume",
        **h1_log
    },

    {
        "analysis":
            "H2 7-day volatility",
        **h2_result
    },

    {
        "analysis":
            "H2 14-day volatility",
        **h2_14d
    },

    {
        "analysis":
            "H3 mean sentiment",
        **h3_result
    },

    {
        "analysis":
            "H3 median sentiment",
        **h3_median
    }
])


robustness_results[
    [
        "analysis",
        "n",
        "spearman_rho",
        "p_value",
        "ci_95_low",
        "ci_95_high"
    ]
].round(4)
```

### Sensitivity and Robustness Interpretation

The sensitivity analysis indicates that the main findings are generally stable under reasonable alternative specifications. For H1, using logarithmic trading volume produces exactly the same Spearman correlation as the original volume measure (\(\rho = 0.4566\), \(p < 0.001\)). This is expected because Spearman correlation is based on ranks, and the logarithmic transformation preserves the ordering of positive trading-volume observations. The H1 conclusion is therefore highly robust to this transformation. For H2, replacing seven-day volatility with fourteen-day volatility strengthens the negative association with network activity from \(\rho = -0.1691\) to \(\rho = -0.2538\). Both relationships remain statistically significant, and the confidence intervals remain below zero. This supports the conclusion that the negative network-volatility relationship is not dependent on the specific seven-day volatility window.

For H3, using median rather than mean daily sentiment reduces the estimated correlation from \(\rho = 0.1925\) to \(\rho = 0.1001\), and the relationship remains statistically non-significant. The confidence interval also continues to include zero. The overall H3 conclusion is therefore unchanged: the available data do not provide convincing evidence that headline sentiment is associated with subsequent XRP returns. Overall, the robustness checks strengthen confidence in the H1 and H2 conclusions, while confirming that the absence of statistical support for H3 is not solely a consequence of the chosen daily sentiment aggregation method.

```python
#XMASI Leave-One-Component-Out Check
xmasi_complete = (
    xmasi_data
    .dropna(
        subset=
        xmasi_components
        + [
            "future_volatility_7d"
        ]
    )
    .copy()
)


xmasi_sensitivity = []


for removed_component in xmasi_components:

    remaining_components = [
        component
        for component
        in xmasi_components
        if component != removed_component
    ]

    alternative_index = (
        xmasi_complete[
            remaining_components
        ]
        .mean(axis=1)
    )

    rho, p_value = stats.spearmanr(
        alternative_index,
        xmasi_complete[
            "future_volatility_7d"
        ]
    )

    xmasi_sensitivity.append({
        "component_removed":
            removed_component,

        "n":
            len(xmasi_complete),

        "spearman_rho":
            rho,

        "p_value":
            p_value
    })


xmasi_sensitivity = pd.DataFrame(
    xmasi_sensitivity
)


xmasi_sensitivity.round(4)
```

### Robustness Evaluation

The sensitivity analysis indicates that the principal hypothesis conclusions are generally stable under reasonable alternative specifications. H1 remains unchanged when logarithmic rather than raw trading volume is used, while H2 remains statistically significant when volatility is measured over fourteen rather than seven days. H3 remains non-significant when median sentiment replaces mean sentiment. These results increase confidence that the main findings are not driven solely by one particular variable definition.

The leave-one-component-out analysis provides a further robustness check for XMASI. Removing each component individually leaves the association with future volatility negative in every case, although the estimated strength varies. The largest change occurs when the trading-volume component is removed, producing \(\rho = -0.3624\), while removing the return component reduces the relationship to almost zero (\(\rho = -0.0256\)). Importantly, none of the alternative index specifications is statistically significant at the 5% level.

The XMASI conclusion is therefore robust in one important respect: neither the complete index nor any leave-one-component-out version provides statistically significant evidence of an association with subsequent seven-day volatility. However, the variation in correlation magnitude indicates that the index remains somewhat sensitive to its composition. Combined with the small sample of 20 complete observations, XMASI should therefore be interpreted as an exploratory descriptive indicator rather than a validated forecasting tool. Overall, the robustness checks strengthen confidence in the findings for H1 and H2, confirm the absence of statistical support for H3, and demonstrate that the limited predictive evidence for XMASI is not dependent on any single component. These analyses assess the stability of the observed associations but do not establish causality or forecasting performance.

# 7. Conclusion

This project investigated how XRP trading activity, XRP Ledger network activity and public media sentiment were associated with XRP market behaviour. A multi-source python workflow was created to acquire, validate, clean, transform and integrate market, blockchain and media data with a reproducible approach to conduct exploratory and inferential statistical analysis. The study confirms some aspects of XRP activity are related to market behavior, with varying degrees of strength, but different directions. Trading volume had the strongest correlation with market activity, and XRP Ledger activity with volatility was less correlated. Conversely, the media-sentiment observations available were not statistically significant with subsequent returns in XRP. The XRP Market Attention and Stress Index (MASI), which serves as a measure of market sentiment, was not found to be predictive of future market volatility. These results should therefore be considered as a statistical association of the observed data, and not as evidence of causal mechanisms, reliable market prediction, or anything else.

## 7.1 Achievement of Aims and Objectives
The main objective of the project was to explore the relationship between the trading activity of XRP tokens, the activity of the XRP Ledger, public media sentiment, and XRP returns and market volatility. This goal was accomplished by developing an integrated analytical framework that integrates three independent data sources: cryptocurrency market data, blockchain network activity, and media coverage of XRP.

The first objective was to obtain relevant data programmatically from independent public sources. CoinGecko was used for XRP market observations, XRP transaction activity was sourced from the Coin Metrics public data archive, and the XRP related media observations were sourced from a public cryptocurrency news archive. Raw data were kept in separate data sets to ensure data provenance and data reproducibility.

The second objective was to develop a reproducible validation and cleaning pipeline. The missing value, duplicate, invalid record, data type and temporal consistency of market, network and media datasets were evaluated separately. Raw/cleaned DataFrames were kept separately, and schema checks, assertions and a targeted test-driven development session were employed to ensure key processing behavior.

The third goal was to design variables which were suitable to the research questions. These encompassed daily and absolute XRP returns, rolling volatility, media attention measures, variables measuring the volume of trading on the XRP Ledger, as well as variables which correlated sentiment with subsequent returns. The three sources were then aggregated on a daily basis, maintaining the missing sentiment on dates when there was no data from the media source.

The fourth and fifth goals were related to exploratory analysis and formal hypothesis testing. The predefined hypotheses were assessed using descriptive statistics, visualisation, distribution diagnostics, Spearman correlation, bootstrap confidence intervals and sensitivity analyses. Lastly, the project built and analyzed the Exploratory XRP Market Attention and Stress Index (XMASI), a multi-source composite measure, rather than focusing on individual relationships.

The overall project was thus successful in terms of its analytical and programming goals, not least because of the clear progression of the data acquisition to the validation of the data, the transformation, integration and statistical interpretation.

## 7.2 Summary of Key Findings

The most compelling result was related to trading volume and the size of price fluctuations in XRP. H1 identified a moderate positive association between daily trading volume and absolute return, with Spearman's \(\rho = 0.4566\) based on 350 paired observations. The relationship was statistically significant with a p value of < 0.001, and the 95% bootstrap CI (0.3658, 0.5448) lay completely above zero. Thus, the null hypothesis was rejected, and the results indicated that days with more trading activity of XRP are correlated with more price changes. The same rank correlation was obtained by performing a sensitivity analysis with the log trading volume, so this result was not sensitive to the transformation of the volume variable.

H2 found that although there was a weak, statistically significant negative correlation, there was no strong relationship between the activity of transactions on the XRP Ledger and the seven-day rolling market volatility. The correlation was \(\rho = -0.1691\), with \(p = 0.0081\) and a 95% confidence interval from -0.2879 to -0.0359. The null hypothesis of no association was therefore rejected, but the small magnitude suggests that activity in the network only accounts for a small part of volatility behaviour. Considering 14 days volatility, the result was again directionally consistent and moderately stronger as \(\rho = -0.2538\) and \(p = 0.0001\). This added weight to the confidence that the correlation between the observed negative correlation was not only with respect to the definition of seven days volatility.

The available evidence is insufficient to support H3. The mean daily headline sentiment was slightly negatively correlated with the next day's XRP return with a correlation coefficient of \(\rho = -0.1925\) and a significance level of \(p = 0.1307\). The 95% confidence interval included 0, (0.4128, -0.0338). The conclusion remained unchanged when median sentiment was used instead of mean sentiment. The available data thus do not give enough evidence on whether there was a systematic relationship between daily headline sentiment and the daily returns of XRP on the subsequent day.

Again, the evidence for a persistent and strong relationship between media sentiment and XRP returns in the exploratory lag analysis was weak. The highest value of the correlation coefficient was observed at lag = -1 day, \(\rho = 0.2222\), but this was not significant (\(p = 0.0800\)) and failed the Bonferroni correction. The strongest association was at a negative lag, which is more a reflection of the possibility that media sentiment is a response to recent market behaviour, than of a fixed association.

Finally, XMASI provided an original method for combining market movement, trading activity, network activity, media attention and negative sentiment within a single standardised indicator. However, its association with subsequent seven-day volatility was weak and non-significant (\(\rho = -0.1519\), \(p = 0.5227\)). High-XMASI observations were also not followed by greater average volatility than other observations. It is therefore more appropriate to consider the index as a descriptive indicator than a validated forecasting indicator.

## 7.3 Evaluation of Approach and Limitations

One of the key advantages of the study is the ability to incorporate both market and blockchain and media information into a single reproducible Python pipeline. Separation of raw, cleaned and processed datasets enhance traceability, explicit schema validation and assertions, and unit testing minimize the potential for processing errors going undetected. The various tools used to transform the data were illustrated, including vectorized Pandas, SQL for relational aggregation and the pre-trained NLP tool VADER, which does not require a labelled training set for XRP. The statistical method was also chosen because of the observed properties of the data, not because of any assumption prior to the observation of the data. Spearman rank correlation was used due to the fact that there was a significant amount of skewness, extreme observations and a significant departure from normality. Bootstrap confidence intervals were also reported with the p-value for consideration of uncertainty and effect size. Sensitivity analysis also showed that the main conclusions for H1, H2 and H3 were not sensitive to any particular reasonable specification for the variables.

Several limitations nevertheless affect the interpretation of the findings. The first is that it is an observational study. Relationships among trading volume, network behavior, sentiment and market returns are not causal in nature and unobserved market-wide variables may affect multiple of the variables. Second, observations are time series and could have serial dependence. This temporal dependence is not explicitly modeled in the standard Spearman correlation and in the ordinary bootstrap procedure. Because of this, the confidence intervals are meant to be used as an approximation of uncertainty, and should not be used as a comprehensive time-series treatment. Thirdly, there is significant variation in the availability of data between data sources. Network data cover a shorter time period, headline sentiment is observed on only 63 data points, while market data give the best daily coverage. The smaller sentiment sample decreases the statistical power and conclusion's strength of H3. Media observations are also subject to the coverage and selection of the source archive, and cannot be taken as representative of all public discussion of XRP.

The sentiment method adds more restrictions. VADER is a transparent, reproducible measure of textual tone but not trained on financial/cryptocurrency language. Short headlines can be ambiguous or contain technical terminology which is not fully understood by a rule-based model. It is therefore important to consider headline sentiment as an approximation of the tone of the media and not as an indicator of investor sentiment. The XRP Ledger transaction-count variable also is activity, not economic. The transaction count is not necessarily correlated with transaction value, number of unique users or number of adoption. Ledger activity can be related to various types.

Lastly, XMASI's evaluation is limited to the 20 observations with full information for all components and future volatility. It was chosen for equal weighting with the intention of making it transparent, not through statistical optimisation, but the small complete case sample does not allow for strong claims for predictive performance. Thus, the negative XMASI result is informative, not that no useful multi-source indicator could be developed, but that it is a judgment on the proposed method.

## 7.4 Implications and Possible Future Research

The results add to the general cryptocurrency literature on Bitcoin and the broader market by demonstrating that the links found in the Bitcoin and general cryptocurrency literature are not necessarily found in the case of XRP. The positive association between XRP trading volume and absolute returns is broadly consistent with previous evidence that trading activity contains information about cryptocurrency price behaviour. Balcilar et al. (2017) identified a nonlinear relationship between Bitcoin trading volume and returns, while Aalborg, Molnár and de Vries (2019) found that trading volume improved models of Bitcoin volatility. Although these studies primarily examined Bitcoin, the present findings indicate that trading activity is also an important dimension of XRP market behaviour. The current results thus support the idea that trading events play a significant role in the dynamics of the cryptocurrency markets, although causal prediction cannot be inferred from the current analysis. This could be furthered in the future through the use of intraday XRP data and the use of models that would detect if the volume of trading increased before price fluctuations and if it increased after price fluctuations.

The connection between the activity of the blockchain and market behavior was not as clear. The activity of transactions on XRP Ledger was negatively associated with seven days volatility, although the correlation was not statistically significant, and the negative link was marginally stronger with fourteen days volatility. There have also been precedents in Bitcoin research that indicates the relationship between blockchain activity and market behaviour. For instance, Koutmos (2018) documented that the Bitcoin transaction activity had bidirectional relationships with returns, whilst Gkillas, Tantoula and Tzagarakis (2021) concluded that Bitcoin transaction activity enhanced out-of-sample forecasts of realised volatility. The negative relationship seen with XRP implies that the correlation between on-chain activity and market conditions can be different across cryptocurrencies and that it does not necessarily have to lead to the same conclusion as the one Bitcoin data can give. XRP trading analysis could then, therefore, differentiate between payment transactions, transactions involving the exchange of XRP, transaction value, active addresses, or other types of activity on the ledger.

The absence of statistical support for H3 is also meaningful when considered alongside previous sentiment research. Guégan and Renault (2021) identified that the sentiment of social media had significant relevance with Bitcoin returns at very high frequencies and not at low frequencies. The non-significant daily relationship found in the current XRP analysis therefore supports the assumption that sentiment data becomes a mere part of the price of cryptocurrencies in relatively short time and that it may become challenging to find it at the one-day timeframe. Future studies might explore XRP sentiment on hourly timeframes or intraday timeframes, and the difference between sentiment gauges based on headlines vs social media. Financial or other cryptocurrency-related language models also may be compared to VADER to see if more context-sensitive NLP techniques capture information that a rule-based approach fails to capture.

The results of the exploratory media-attention and XMASI findings also offer a valuable contrast to the existing evidence. Higher investor attention, as gauged by social-media activity, was linked to higher volatility in cryptocurrencies, with the possibility of it potentially holding information about future volatility, according to Al Guindy (2021). For the current research, however, XMASI had no statistically significant correlation with the following 7-day volatility of XRP. This variation could be due to the significantly smaller number of complete cases used in the analyses, to the different type of attention measure used (frequency of news articles vs. social-media activity), or to the inclusion of multiple conceptually different variables in a single equal-weighted index in XMASI. As such, the findings do not necessarily suggest that multi-source indicators are less effective, but rather that they need to be validated on much larger data sets for a predictive claim to be made.

Future research may then be able to use a longer synchronised sample to develop XMASI and to test alternative weighting schemes with data that have not been included in the development of the index. Out-of-sample testing would be especially relevant, in order to avoid the selection of weights that only fit the sample. In this study, temporal dependence is captured in the correlation structure, but can be handled more explicitly using more advanced time-series techniques, such as vector autoregression, autoregressive volatility models, and block-bootstrap methods. Overall, the comparisons with the literature indicate that, as with many other cryptocurrencies, trading activity is the link with the most consistency between the present XRP results and previous studies on cryptocurrencies, while the relationship with the network and with the sentiment seems to depend more on the analyzed asset, the measure used, and the temporal frequency considered. It serves as a reminder that empirical relationships, which are usually more relevant to the Bitcoin market, do not apply to the cryptocurrency market in general, but rather the market for XRP specifically.
